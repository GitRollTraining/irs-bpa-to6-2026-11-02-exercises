**From the slides:** Slide 5, "A one-line task is a wish: the agent guesses at every gap"; Slide 7, "A directing prompt has four parts; only one checks the result"; Slide 10, "Which done-when could someone who did not watch check?".

**Before you start:** You do not write any code. To run a cell, click it and press Shift+Enter.
- A *coding agent* is an AI that edits files and runs commands. *pytest* runs *tests*: small checks of the code. A *done-when* is a command and what it must show.
- A line that starts with `!` is a real command, run on a practice project; text after `#` is only a note.

**Done when:** the last cell prints `Done:`.
