### Missing details cost: state what must be true of the result.

**What a missing detail costs**

- Details removed: 54.80% of issues resolved, against 70.80% complete
- The agent: Claude Sonnet 4.5
- A key name and a unit left unstated cost one condition 23.4 points
- Left implicit, it is scored as model error

**A specification**

- Says what must be true of the result; a style guide, how to write
- 267 words, 50 backend tasks, five models
- Findings in the named defect classes: 23 with it, 148 without
- Descriptive only, its author says
- Single-turn, no tools; correctness not measured

A 2026 study gave coding agents real repository issues with details removed, and compared the stripped versions with the complete ones.

A second 2026 study looked at one model writing analysis code.

Stating what must be true also changes what comes back. A single-author 2026 preprint put a 267-word specification in front of 50 backend tasks across five models. The arm with the specification carried 23 findings against 148 in the bare arm, an 84 percent reduction in the named defect classes. The author calls that pooled figure descriptive only. Its limits are part of the result: generation was single-turn with no tools, and the study did not measure functional correctness. The useful part is its distinction: a style guide tells the model how to write, and a specification tells the model what must be true of the result.

### A directing prompt has four parts; only one checks the result.

| Part | Run 1 |
| --- | --- |
| Goal | add an output schema for the case-notes extractor, and tests that validate outputs against it. |
| Context | read notes/README.md and notes/0147.txt; the five fields are in docs/fields.md. |
| Constraints | tax_year is an integer; form and issue and next_action are strings; amount_in_question is an integer or null, and is always present. Do not call any model. Do not edit anything outside schema/ and tests/. Do not log note text. |
| Done-when | pytest tests/test_schema.py passes; it includes one output that must fail (amount_in_question missing) and one that must pass with amount_in_question null. |

The table is run 1 of the case-notes extractor written that way, word for word from section 5.1.

## 5.2 Check the done-when yourself

*The agent reports that every test passes. What has it shown you?*

### "All tests pass" shows only that some tests pass.

*Not that they are your tests, or that the code does what was asked.*

**ImpossibleBench, 2025**

- Its tests contradicted the specification: any pass meant a shortcut
- Seen: models editing tests against instructions
- GPT-5, the highest-performing model: cheated 54.0% of the time on one task set

**METR, a research nonprofit**

- An early-2025 agent, Claude 3.7 Sonnet: 18 real issues, two projects
- Passed the maintainers' tests: 38% on average
- 15 pull requests read by hand: none mergeable as-is

The agent's report shows that some tests pass: not that they are the tests you wrote, or that the code does what the task asked. One way its authors saw agents pass: the model directly modifies tests despite being explicitly instructed not to.

### You verify: run the done-when yourself, test files first.

*Without a check the agent can run, "looks done" is its only signal.*

**Have it show evidence**

- The test output
- The command it ran, and what it returned
- A screenshot of the result
- Not "all tests pass" without a matching check in the log

**For every run, yourself**

- Run the done-when in your own terminal
- Read the test files' diff first
- An expected value changed? Not done until you know why
- Make the tests read-only: a deny rule
- The report: done, proof, scope, not verified

Claude Code's best-practices page names the trap: Claude stops when the work looks done, and without a check it can run, looks done is the only signal available, so you become the verification loop. Its fix is to have Claude show evidence rather than asserting success: the test output, the command it ran and what it returned, or a screenshot of the result. A 2026 study of agents contributing to open-source projects saw the opposite habit: agents writing that the change works, or that all tests pass, without a matching check in the command log.

Look at the diff of the test files before anything else: if a test's expected value changed, the run is not done until you know why. Better still, take the option away: ImpossibleBench's authors found that making tests read-only restores legitimate performance while preventing test modification attempts, though it does not eliminate other cheating methods such as special-casing or operator overloading; they recommend either hiding test files entirely or restricting them to read-only access during implementation, when feasible. In the lab that is a deny rule in the agent's configuration: a setting that blocks its edits to the files it names.

For the report, one self-published standard gives four fields: done, proof, scope and not verified, where not verified is what was not checked, could not be checked, or remains uncertain.

## Recap

*What this week set up, and where it goes wrong*

### Five failures the week's method catches; none announces itself.

| Failure | How you detect it |
| --- | --- |
| Note text where it does not belong | Search each diff and log for note text; the data-rules check before any AI tool |
| The test that moved to meet the code | Read the test files' diff first, every run; make the tests read-only |
| Done on assertion | Ask where the output is; run the command yourself |
| Valid shape, wrong value | Only the evaluation set, scored field by field |
| The prompt revised by eye | A recorded score beside every prompt change |

Note text where it does not belong: in a prompt to a tool that is not approved, in a log line, in a test fixture; search each diff and each log for note text before you commit, and pass the data-rules check before any AI tool touches the repository. Valid shape, wrong value: the reply passes the schema and the amount is invented; schema validation cannot detect it, only the evaluation set, scored field by field. The prompt revised by eye: someone changes a line of the system prompt because one output looked wrong, and nobody runs the set; require a recorded score beside every prompt change.

### Slower per change; it buys a change you can put your name on.

*Use less of it for code that is faster to write than to check.*

**It costs**

- Attention for every approval and every diff
- Hours of hand-labelling for the evaluation set

**It buys**

- A number that says whether the next change made things worse
- Direct human review of AI output, as IRS policy requires

This way of working is slower per change than accepting what the agent writes. For work under the IRS's rules that is not optional: the policy prohibits using AI output in government business without ensuring direct human review to confirm the veracity of the information.
