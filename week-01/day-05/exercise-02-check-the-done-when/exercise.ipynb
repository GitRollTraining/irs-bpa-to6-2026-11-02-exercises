# Exercise 2 · Check the done-when yourself

A *done-when* is a command and what it must show; you run one yourself.

**Goal:** Run the done-when yourself; the agent changed a test, so undo its work.

**Why it matters:** An agent can make tests pass by changing them.

**From the slides:** Slide 14, "'All tests pass' shows only that some tests pass"; Slide 16, "You verify: run the done-when yourself, test files first".

**Before you start:** You do not write any code. To run a cell, click it and press Shift+Enter.
- A *coding agent* is an AI that edits files and runs commands. *pytest* runs *tests*: small checks of the code. *Git* tracks file changes.
- A line that starts with `!` is a real command, run on a practice project; text after `#` is only a note.
- `%` lines are notebook helpers, not Git: `%step 2` resets the project to Step 2's start; each step ends with an OK or PROBLEM line.

**What the check measures:** the last run of Steps 2 and 3.

**Time:** about 10 minutes.

**Done when:** the last cell prints `Done:`.

## The situation

You have a small Git project: `total.py` adds up the prices in `groceries.txt`, and `test_total.py` holds the tests. `total.py` stops with an error on `bread`, which has no price. You asked a coding agent: "Fix it. Change only `total.py`. Done-when: `python3 -m pytest -q` passes." It replies "Done. All 3 tests pass." (made up: no real agent here).

**Run the next cell first** (again if you see `UsageError`): it builds the project in `practice/groceries` and prints nothing.

```python
%run steps.py
```

## Step 1 · Run the done-when yourself

"All tests pass" does not prove the code is right (Slide 14). Your done-when, `python3 -m pytest -q`, runs the tests quietly: `.` is a passed test; `[100%]` shows progress. `git status --short` marks each changed file `M`.

Run the cell; it has no TODO line. **Each run starts from the agent's changes.** The last line starts with `PROBLEM: 3 tests pass, but the agent changed test_total.py`.

```python
%step 1
!python3 -m pytest -q
!git status --short
```

## Step 2 · Read the test file's diff

If a test now expects another number, the work is not done until you know why (Slide 16). `--stat` only counts changed lines. `git diff test_total.py` shows them. Under the `@@` line, `-` marks an old line and `+` a new one.

Run the cell. Before your edit, the last line starts with `PROBLEM: this does not show test_total.py's changed lines`. On the TODO line, delete `--stat`:

`!git diff test_total.py`

Run the cell again. After your edit, the last line starts with `OK: you found the cheat`.

```python
%step 2
!git diff --stat test_total.py    # TODO: delete --stat
```

## Step 3 · Undo the agent's work

`git restore` removes the changes in the files you list.

Run the cell. Before your edit, the last line starts with `PROBLEM: the agent's change is still in test_total.py`. On the TODO line, type a space and `test_total.py` right after `total.py`:

`!git restore total.py test_total.py`

Run the cell again. After your edit, `git status --short` lists nothing and the last line starts with `OK: both files are back`.

```python
%step 3
!git restore total.py    # TODO: add test_total.py
!git status --short
```

Run the cell. Expect `Done:`; a `Not yet:` line names the step to redo.

## What you learned

- "All tests pass" does not prove the code is right: run the done-when yourself, `python3 -m pytest -q`, then `git status --short` (Slides 14 and 16).
- Read the test file's diff first, with `git diff test_total.py` (Slide 16); if a test changed, undo the agent's work with `git restore`.
