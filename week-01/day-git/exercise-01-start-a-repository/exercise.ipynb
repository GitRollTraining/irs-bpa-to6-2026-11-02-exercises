## Step 2 · Stage the files

`git add` copies the files you name into the *staging area*, the draft of your next commit. `git status` then lists them under `Changes to be committed`; a file you have not added stays under `Untracked files`. `git add .` would stage every file in the folder at once.

Run the cell. Before your edit, `groceries.txt` stays under `Untracked files`; the last line starts with `PROBLEM: groceries.txt is not staged`. On the TODO line, type `groceries.txt` after `meals.txt`:

`!git add meals.txt groceries.txt`

Run the cell again. After your edit, both files are under `Changes to be committed`; the last line starts with `OK: both files are staged`.

```python
%step 2
!git add meals.txt    # TODO: type groceries.txt after meals.txt
!git status
```

## Step 3 · Commit, and read the log

`git commit` saves what you staged as a commit; `-m "…"` is its message, which says what the commit does. `git log` lists the commits, newest first, each with its author and date.

Run the cell. Before your edit, Git prints `Aborting commit due to empty commit message`, and `git log` prints `fatal: your current branch 'main' does not have any commits yet`; the last line starts with `PROBLEM: no commit was made`. On the TODO line, type `Start the meal plan` between the quotes:

`!git commit -m "Start the meal plan"`

Run the cell again. After your edit, Git prints `2 files changed`, and `git log` shows your commit; the last line starts with `OK: your first commit holds both files`.

```python
%step 3
!git commit -m ""    # TODO: type Start the meal plan between the quotes
!git log
```
