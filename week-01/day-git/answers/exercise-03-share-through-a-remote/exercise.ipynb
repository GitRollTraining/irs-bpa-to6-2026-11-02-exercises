## Step 1 · Add the remote

`git remote add origin <address>` saves the address under the name `origin`; Git does not check the address until you push or pull. `git remote -v` lists each remote with its address, once to fetch from and once to push to.

Run the cell. Before your edit, `origin` points to `URL`; the last line starts with `PROBLEM: origin points to URL`. On the TODO line, change `URL` to `../github/meal-plan.git`:

`!git remote add origin ../github/meal-plan.git`

Run the cell again. After your edit, `git remote -v` lists `../github/meal-plan.git`; the last line starts with `OK: origin points to the shared repository`.

```python
%step 1
!git remote add origin ../github/meal-plan.git
!git remote -v
```

## Step 2 · Push your commits

`git push -u origin main` sends the commits on your branch `main` to `origin`. `-u` also links your `main` to `origin/main`, so a plain `git push` or `git pull` later knows where to go. You push a branch you have: `git branch` lists yours. Some guides name the first branch `master`; this project's is `main`.

Run the cell. Before your edit, Git prints `error: src refspec master does not match any`; the last line starts with `PROBLEM: nothing was pushed`. On the TODO line, change `master` to `main`:

`!git push -u origin main`

Run the cell again. After your edit, Git prints `main -> main`; the last line starts with `OK: the shared repository holds your 2 commits`.

```python
%step 2
!git branch
!git push -u origin main
```

## Step 3 · Pull your housemate's commit

Since your push, your housemate has added a Friday dinner and pushed it. `git push` sends your commits, and Git rejects it while `origin` holds a commit you do not have. `git pull` gets the new commits from `origin` and adds them to your branch. You have no new commits of your own, so the pull moves your `main` forward to their commit: Git calls this a fast-forward.

Run the cell. Before your edit, Git prints `! [rejected]` and a hint to use `git pull`, and `git log` lists your 2 commits; the last line starts with `PROBLEM: Git rejected your push`. On the TODO line, change `push` to `pull`:

`!git pull`

Run the cell again. After your edit, Git prints `Fast-forward`, and `git log` lists 3 commits, `Add a Friday dinner` first; the last line starts with `OK: pull brought your housemate's commit`.

```python
%step 3
!git pull
!git log --oneline
```
