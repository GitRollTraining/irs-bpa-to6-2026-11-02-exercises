## Step 1 · Make a branch, and move to it

Run the cell. Before your edit, the log of `main` lists 2 commits, without yours; the last line starts with `PROBLEM: this is main's history`. On the TODO line, change `main` to `veggie-week`:

`!git log --oneline veggie-week`
