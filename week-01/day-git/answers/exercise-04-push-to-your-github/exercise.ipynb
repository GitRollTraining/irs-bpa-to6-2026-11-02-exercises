## Step 1 · Clone your repository

`git clone <address>` copies a repository from GitHub into a new folder named after it, with its whole history, and saves the address as `origin`. This cell runs in `practice/`.

Run the cell. Before your edit, Git prints `fatal: repository 'URL' does not exist`; the last line starts with `PROBLEM: nothing was cloned`. On the TODO line, change `URL` to your repository's address, for example:

`!git clone https://github.com/your-name/your-repo.git`

Run the cell again. After your edit, Git prints `Cloning into 'your-repo'...`, and for an empty repository a warning that it is empty; the last line starts with `OK: your repository is cloned`.

```python
%step 1
!git clone https://github.com/your-name/your-repo.git
```

## Step 2 · Add a file

The cell writes a new file, `hello.txt`, in your clone. `git add` stages it for your next commit.

Run the cell. Before your edit, Git prints `fatal: pathspec 'FILE' did not match any files`, and `git status` lists `hello.txt` under `Untracked files`; the last line starts with `PROBLEM: nothing was staged`. On the TODO line, change `FILE` to `hello.txt`:

`!git add hello.txt`

Run the cell again. After your edit, `git status` lists `hello.txt` under `Changes to be committed`; the last line starts with `OK: hello.txt is staged`.

```python
%step 2
!echo "Hello from the Git exercises" > hello.txt
!git add hello.txt
!git status
```

## Step 3 · Commit

`git commit -m "…"` saves what you staged as a commit, signed with the name and email from your Git settings.

Run the cell. Before your edit, Git prints `Aborting commit due to empty commit message`; the last line starts with `PROBLEM: no commit was made`. On the TODO line, type `Add hello.txt` between the quotes:

`!git commit -m "Add hello.txt"`

Run the cell again. After your edit, Git prints `1 file changed`, and `git log` lists your commit first; the last line starts with `OK: your commit holds hello.txt`.

```python
%step 3
!git commit -m "Add hello.txt"
!git log --oneline -3
```

## Step 4 · Push

`git push` sends your commit to `origin`, your repository on GitHub, signed in as you. `git status` then compares your branch with GitHub's.

Run the cell. Git prints your branch's name twice, such as `main -> main`, and `git status` says your branch is up to date with `origin`; the last line starts with `OK: your commit is on GitHub`.

```python
%step 4
!git push
!git status
```

Run the cell. Before your edit, Git prints `fatal: 'URL' does not appear to be a git repository`; the last line starts with `PROBLEM: nothing was pushed`. On the TODO line, change `URL` to your repository's address, the one you cloned:

`!git push https://github.com/your-name/your-repo.git veggie-week`
