## The situation

Step 1 writes a small Python project into a folder called `practice`: `textstats.py` counts words and sentences and finds the average word length, and `test_textstats.py` tests it. One function does not do what its docstring says, and Codex will find it for you.

## Step 1 · Write the practice files

Run the cell. It writes two files into `practice/`, a folder beside this notebook.

Expect two lines that start with `Wrote`, or with `Kept` if the files were there already.

````python
# Step 1: write the practice files into practice/, keeping any that are already there
import pathlib

PRACTICE = pathlib.Path("practice")
STARTING_FILES = {
    "textstats.py": '''\
"""Simple statistics about a piece of text."""


def word_count(text: str) -> int:
    """Return the number of words in text."""
    return len(text.split())


def sentence_count(text: str) -> int:
    """Return the number of sentences: each one ends with '.', '!' or '?'."""
    return text.count(".") + text.count("!")


def average_word_length(text: str) -> float:
    """Return the mean length of the words, punctuation included; 0.0 for no words."""
    words = text.split()
    if not words:
        return 0.0
    return sum(len(word) for word in words) / len(words)
''',
    "test_textstats.py": '''\
from textstats import average_word_length, sentence_count, word_count


def test_word_count():
    assert word_count("The cat sat.") == 3


def test_sentence_count():
    assert sentence_count("Hi. Are you there? Yes!") == 3


def test_average_word_length():
    assert average_word_length("Hi there.") == 4.0
''',
}
PRACTICE.mkdir(exist_ok=True)
for name, text in STARTING_FILES.items():
    path = PRACTICE / name
    if path.exists():
        print(f"Kept practice/{name}")
    else:
        path.write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

### Learn: The sandbox and approvals

Two settings control Codex:

- The **sandbox** limits what Codex's commands can reach. `read-only` lets Codex read files but not change them. `workspace-write` also lets it edit files in its folder.
- The **approval policy** says when Codex stops to ask you. `on-request` asks before Codex does something the sandbox does not allow.

On read-only with on-request, the docs say: "Codex can read files and run commands within the read-only sandbox. Actions outside the sandbox can require approval." ([Agent approvals & security](https://learn.chatgpt.com/codex/agent-approvals-security), "Common sandbox and approval combinations")

**Your rule today:** Codex usually reads without asking. If it asks to run a command, read it first. If the command only shows, lists or searches files, such as `ls`, `cat`, `sed -n`, `nl`, `rg` or `grep`, approve that command once: "When an approval offers different scopes, such as approving once or for the session, choose the narrowest scope that lets the task continue" ([Sandbox](https://learn.chatgpt.com/codex/sandboxing), "How permissions work"). Say no to anything else, such as running the tests or changing a file.

## Start over (only if you need it)

This cell writes both practice files again as Step 1 wrote them. It does nothing until you set `START_OVER = True`.

Expect `Start over is off: no file changed.` With `START_OVER = True`, expect two lines that start with `Rewrote`; then set it back to `False`.

````python
# Start over: rewrite both practice files as Step 1 wrote them, only when START_OVER is True
START_OVER = False  # set to True to start over
if START_OVER:
    PRACTICE.mkdir(exist_ok=True)
    for name, text in STARTING_FILES.items():
        (PRACTICE / name).write_text(text, encoding="utf-8")
        print(f"Rewrote practice/{name}")
else:
    print("Start over is off: no file changed.")
````

## Check your work

Run the cell. Expect three lines that start with `ok:`, then `Done:`. If the last line starts with `Not yet:`, do what it says, then run this cell again.

````python
# Check your work: the practice files are unchanged, and BUGGY_FUNCTION is right
import re


def problem() -> str:
    """Print each check that passes; return the first thing to fix, or "" if none."""
    for name, text in STARTING_FILES.items():
        path = PRACTICE / name
        if not path.is_file():
            return f"practice/{name} is missing. Run the Step 1 cell."
        if path.read_text(encoding="utf-8", errors="replace") != text:
            return (f"practice/{name} changed: say no if Codex asks to change it, and do not edit it "
                    "yourself. To put it back, set START_OVER = True in the Start over cell and run it.")
        print(f"ok: practice/{name} is unchanged")
    if BUGGY_FUNCTION is None:
        return "BUGGY_FUNCTION is still None. In Step 8, put the name Codex gave in quotes, then run that cell."
    if not re.search(r"\bsentence_count\b", str(BUGGY_FUNCTION).lower()):
        return (f"{BUGGY_FUNCTION!r} is not the function whose code misses part of its docstring. "
                "Compare Codex's answer in Step 6 with the Step 1 cell, then fix and run Step 8.")
    print("ok: BUGGY_FUNCTION names sentence_count")
    return ""


try:
    found = problem()
except NameError as error:  # a cell above has not run, or it stopped with an error
    found = f"{error.name} is not set. Run each cell from Step 1 down; in Step 8, put the name in quotes."
except Exception as error:
    found = f"the check stopped with {type(error).__name__}. Run each cell from Step 1 down, then this one."
print(f"Not yet: {found}" if found else 'Done: Codex changed no file, and you found sentence_count, which misses "?".')
````
