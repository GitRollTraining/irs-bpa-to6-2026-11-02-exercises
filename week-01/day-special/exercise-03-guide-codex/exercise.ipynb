## Step 1 · Write the practice files

Run the cell. It writes four files into `practice/`, beside this notebook: `converter.py`; `test_converter.py`; `pytest.ini`, which keeps pytest's own files in `practice/`, where Codex may write; and `.gitignore`, which keeps them out of Git. It never overwrites a file.

Expect four lines that start with `Wrote practice/`, or `Kept` for files already there.

````python
# Step 1: write the practice files, keeping any that already exist
from pathlib import Path

PRACTICE = Path("practice")
FILES = {
    "converter.py": '''"""Convert between common units."""


def km_to_miles(km: float) -> float:
    """Return km in miles, rounded to 2 places."""
    return round(km * 0.621371, 2)


def celsius_to_fahrenheit(celsius: float) -> float:
    """Return celsius in degrees Fahrenheit, rounded to 1 place."""
    return round(celsius * 9 / 5 + 32, 1)
''',
    "test_converter.py": '''from converter import celsius_to_fahrenheit, km_to_miles


def test_km_to_miles():
    assert km_to_miles(10) == 6.21


def test_celsius_to_fahrenheit():
    assert celsius_to_fahrenheit(100) == 212.0
''',
    "pytest.ini": "[pytest]\n",
    ".gitignore": "__pycache__/\n.pytest_cache/\n",
}

PRACTICE.mkdir(exist_ok=True)
for name, text in FILES.items():
    if (PRACTICE / name).exists():
        print(f"Kept practice/{name}: it is already there.")
    else:
        (PRACTICE / name).write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

## Start over (only if you need it)

The cell changes nothing while `START_OVER = False`. Set it to `True` and run the cell: it deletes `AGENTS.md` and `summary.md` and rewrites the Step 1 files; your Git history stays. Then set it back to `False` and go to Step 3, where Git may print `re-init` or `nothing to commit`: both are fine.

Expect `Nothing changed.`

````python
# Start over: only if START_OVER is True, rewrite the files as Step 1 wrote them
START_OVER = False

if not START_OVER:
    print("Nothing changed.")
elif "FILES" not in globals():
    print("Run the Step 1 cell first, then this cell.")
else:
    PRACTICE.mkdir(exist_ok=True)
    for name in ("AGENTS.md", "summary.md"):
        (PRACTICE / name).unlink(missing_ok=True)
    for name, text in FILES.items():
        (PRACTICE / name).write_text(text, encoding="utf-8")
    print("Rewrote the practice files. Now set START_OVER back to False.")
````

## Check your work

Run the cell; it reads only `practice/`.

Expect five `ok:` lines, then a line that starts with `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: read practice/, try kg_to_pounds, and run the tests
import contextlib
import io
import re
import runpy
import sys
from pathlib import Path

import pytest

PRACTICE = Path("practice")


def read(name: str) -> str:
    """Return practice/<name> as text, or "" if it is missing."""
    path = PRACTICE / name
    return path.read_text(encoding="utf-8") if path.is_file() else ""


def first_problem() -> str:
    """Print ok for each passing check; return the first problem, or "" if none."""
    for words in ("python -m pytest -q", "docstring"):
        if words not in read("AGENTS.md").lower():
            return f"practice/AGENTS.md has no line with {words}. Write it (Step 4), then start Codex (Step 5)."
    print("ok: AGENTS.md has both rules")
    code = runpy.run_path(str(PRACTICE / "converter.py"))  # a fresh run of the file
    if not getattr(code.get("kg_to_pounds"), "__doc__", None):
        return "converter.py has no kg_to_pounds with a docstring. Ask Codex for it (Step 6)."
    for name, value, right in [("kg_to_pounds", 10, 22.05), ("kg_to_pounds", 0, 0.0),
                               ("km_to_miles", 10, 6.21), ("celsius_to_fahrenheit", 100, 212.0)]:
        got = code[name](value)
        if got != right:
            return f"{name}({value}) gives {got!r}, not {right}. Ask Codex to fix it (Step 6)."
    print("ok: kg_to_pounds has a docstring and works")
    if not re.search(r"kg_to_pounds\(0", read("test_converter.py")):
        return "no test calls kg_to_pounds(0). Ask for one after codex resume --last (Step 8)."
    print("ok: a test calls kg_to_pounds(0)")
    for name in ("converter", "test_converter"):  # so pytest reads the files again
        sys.modules.pop(name, None)
    with contextlib.redirect_stdout(io.StringIO()) as out, contextlib.redirect_stderr(io.StringIO()):
        exit_code = pytest.main(["-q", "--color=no", "-p", "no:cacheprovider", str(PRACTICE)])
    if exit_code != 0:
        last = out.getvalue().strip().splitlines()[-1]
        return f"the tests fail ({last}). Run python -m pytest -q in practice/ to see why."
    print("ok: the tests pass")
    if "pound" not in read("summary.md").lower():
        return ('practice/summary.md is missing or does not name kg_to_pounds. Run Step 9 again; '
                'after Step 10, ask about "the last commit".')
    print("ok: summary.md holds the codex exec answer")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Ask Codex to fix it, or run Step 1."
print(f"Not yet: {problem}" if problem else
      "Done: Codex followed your AGENTS.md unasked, the tests pass, and codex exec saved its answer.")
````
