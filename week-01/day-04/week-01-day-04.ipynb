## 4.2 Fetch, pull and push

*Where is the copy your reviewer reads, and what if someone pushed first?*

### A remote is a destination, so the data rules apply to it.

**A remote**

- A version of your project hosted somewhere else
- Cloning set one up: origin, the server you cloned from
- git remote -v lists each, with its fetch and push URLs
- Before the first push, run it: is origin the lab's remote?

**A push**

- Sends your commits to the URL origin holds
- A clone gets nearly all the data the server has
- So what you push is in every later clone
- Needs write access: a credential, part of the agent's tool access

Pro Git: "Remote repositories are versions of your project that are hosted on the Internet or network somewhere." Cloning set one up for you, called origin, which Pro Git describes as "the default name Git gives to the server you cloned from".

Pushing sends your commits to whatever URL that name holds, and whoever clones from there gets more than the latest files: "Git receives a full copy of nearly all data that the server has." Everything you push is in every later clone, which is why section 1.1 calls committed data so hard to get back.

A push works only if you have write access to the server. In this week's lab you push, after you have compared what is going.

GitHub has a specific way of contributing code to other projects, called pull requests". A pull request is that proposal: commits on the remote, offered for someone to read before they join the main line. Branches, pull requests that a teammate reviews, and merges on a shared team repository are week 2's subject.

### A fetch only downloads; a pull also tries to merge into yours.

*To see what changed before it touches your work: fetch, compare, then merge.*

| Command | What it does | Your work |
| --- | --- | --- |
| git fetch | Retrieves objects and references from a remote | Not modified: it only downloads |
| git pull | git fetch, then git merge: incorporates what came in | Tries to merge into it, automatically |
| git push | Sends objects, and updates the remote reference | Sent only if nobody pushed in the meantime |

In the MIT notes' one-line summaries, fetch is "retrieve objects/references from a remote", push is "send objects to remote, and update remote reference", and pull is "same as git fetch; git merge".

Pro Git draws the line between fetch and pull: "It’s important to note that the git fetch command only downloads the data to your local repository — it doesn’t automatically merge it with any of your work or modify what you’re currently working on." Pull does both: "Running git pull generally fetches data from the server you originally cloned from and automatically tries to merge it into the code you’re currently working on." So when you want to see what changed on the remote before it touches your work, fetch, compare (section 4.3), and only then merge.

*Where each command moves work. origin is the remote you cloned from.*

<img src="figures/fetch-pull-and-push.png" width="864" alt="Three places as columns: your work, your local repository, and origin, the remote. git fetch, in orange, is an arrow from origin to your local repository: it downloads, and your work is untouched. git pull is an arrow from origin all the way to your work: it fetches, then tries to merge into your work. git push is an arrow from your local repository to origin: it sends your commits.">

*A fetch only downloads; a pull also tries to merge into the work you are on.*

## 4.3 Compare before and after you commit

*git status says the agent changed the source file. What did it change?*

### Run git diff --staged: the plain form misses what is staged.

**A diff shows lines**

- git status names files; git diff shows the lines added and removed
- What it compares depends on how you call it
- HEAD: the commit you are on; HEAD^: the one before

**The trap: the plain form**

- Shows only changes still unstaged
- Empty if nothing changed, or if all is staged
- Run git diff --staged before every commit
- It is exactly what the commit will hold
- Search it for note text before it goes in

Pro Git: "git diff shows you the exact lines added and removed — the patch, as it were." The Git reference manual's own one-line description of the command is "Show changes between commits, commit and working tree, etc".

*The four forms of the command. The manual calls the staging area the index.*

<img src="figures/four-forms-of-git-diff.png" width="864" alt="A table of four commands. git diff compares the working directory against the staging area: what has changed that you have not staged. git diff --staged compares the staging area against the last commit: exactly what the next commit will hold, marked in orange. git diff HEAD compares the working directory against the last commit: everything since the last commit, staged or not. git diff HEAD^ HEAD compares the commit before the last against the last: what the last commit changed.">

*git diff --staged shows exactly what the next commit will hold; plain git diff does not.*

The same command compares the tips of two branches, "Changes between the tips of the topic and the master branches." Week 2 uses that form to compare a teammate's work with the main line.

*Source: git-scm.com, git-diff Documentation (Git 2.56.0); Pro Git, 2nd edition, 2.2 Recording Changes to the Repository*

### Start wide, then point --exit-code at HEAD for the tests.

*--exit-code, a check a script can run: 1 if there were differences, 0 if none.*

| The agent's test edit | git diff --exit-code -- <tests> | git diff --exit-code HEAD -- <tests> |
| --- | --- | --- |
| Not staged yet | 1: caught | 1: caught |
| Already staged | 0: missed | 1: caught |

Start wide. --name-status lists files before lines: "Show only the name(s) and status of each changed file." For the extractor that list is the first check after every run, because a file outside what the directing prompt allowed is a finding before you read a single line of it. Then read the test files, then the rest; day 5 comes back to why the tests come first.

--exit-code turns a diff into a check a script can run: "That is, it exits with 1 if there were differences and 0 means no differences." Point it at the last commit, not at the staging area: for the form that compares your working tree with a named commit, the manual says "You can use HEAD to compare it with the latest commit".

## 4.4 Which kind of AI coding tool

*Completion, a chat panel, an agent in a terminal: one tool, three faces?*

### Three kinds of tool, and they need different checks.

*The difference that matters: what the tool can do without you.*

**In MIT's lecture**

- AI-assisted coding: autocomplete and inline chat
- Coding agents: conversational AI models with tools
- Tools: reading and writing files, web search, shell commands

**A study of prompt injection**

- Agentic coding editors have system privileges
- The risk shifts to system security
- It tested GitHub Copilot in VS Code
- So one product can be several kinds, by its mode

The MIT lecture separates them: the MIT course's previous lecture covered "some AI-assisted coding like autocomplete and inline chat", while "Coding agents are conversational AI models with access to tools such as reading/writing files, web search, and invoking shell commands." "They live either in the IDE or in standalone command-line or GUI tools." The IDE is the editor you write code in.

A study of instructions planted in the files such editors read, the prompt injection of section 1.3, puts it sharply: "agentic AI coding editors have system privileges, and attacks can directly compromise the host machine.

*The kinds follow MIT's lecture; the checks are this course's.*

<img src="figures/three-kinds-of-coding-tool.png" width="864" alt="A table of three kinds of tool. Completion suggests the next lines as you type, reaches what it sends to the model, and you check every line before you accept it. Chat answers in a panel while you copy code in and out, reaches what you paste into it, and you check the data you paste and the code you take. An agent reads, edits and runs commands in a loop, reaches whatever its sandbox and credentials allow, and you check its sandbox, approvals and tool access first, then its result against the done-when, marked in orange.">

*For an agent, the check starts before its result: its sandbox, approvals and tool access come first.*

Completion suggests the next lines as you type; you check every line before you accept it. Chat answers in a panel and you copy code in and out; you check the data you paste and the code you take.
