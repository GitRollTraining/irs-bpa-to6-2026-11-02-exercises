## Step 1 · Ignore the log

A log can hold private details, so it never goes in a commit (Slide 9). `git status --short` lists each changed file: `M` changed, `??` new, never committed. `echo "…" >>` adds a line to `.gitignore`, which then shows `M`.

Run the cell. Before your edit, the last line starts with `PROBLEM: prices.log is not ignored`. On the TODO line, type `prices.log` between the quotes:

`!echo "prices.log" >> .gitignore`

Run the cell again. After your edit, the last line starts with `OK: prices.log is ignored`.

```python
%step 1
!echo "prices.log" >> .gitignore
!git status --short
```

## Step 2 · Stage only what you mean

`git add` copies only the files you name into the staging area (Slide 6). In `git status --short`, an `M` in column 1 means staged; in column 2, not staged.

Run the cell. Before your edit, the last line starts with `PROBLEM: budget.txt is staged`. On the TODO line, delete `budget.txt`:

`!git add .gitignore groceries.txt meals.txt`

Run the cell again. After your edit, the last line starts with `OK: 3 changes are staged`.

```python
%step 2
!git add .gitignore groceries.txt meals.txt
!git status --short
```

## Step 3 · Commit without -a

`git commit` saves what you staged; `-m "…"` is the commit's message. The cell lists your changes, then commits. `-a` first stages every file marked `M`, so `budget.txt` goes in too (Slide 8).

Run the cell. Before your edit, Git prints `4 files changed`; the last line starts with `PROBLEM: -a staged budget.txt too`. On the TODO line, delete `-a`:

`!git commit -m "Add eggs and a Tuesday dinner"`

Run the cell again. After your edit, Git prints `3 files changed`; the last line starts with `OK: your commit holds the 3 staged changes`.

```python
%step 3
!git status --short
!git commit -m "Add eggs and a Tuesday dinner"
```
