## Step 1 · Check origin, then fetch

`git remote -v` shows where your data goes: check that it prints `../shared` (Slide 15). Git rejects your push (`[rejected]`) if origin holds a commit your work does not have; this keeps your housemate's work safe (Slide 19). `git fetch` only downloads: origin's `main` goes into your `origin/main` (`main       -> origin/main` below); your work, `main`, is unchanged (Slide 16).

Run the cell. Before your edit, the last line starts with `PROBLEM: Git rejected your push`. On the TODO line, change `push` to `fetch`:

`!git fetch`

Run the cell again. After your edit, the last line starts with `OK: fetch only downloads`.

```python
%step 1
!git remote -v
!git fetch
```

## Step 2 · Read what came in

Read their commit before you merge it (Slide 16). `git show --stat` shows the commit's ID, author, date, message, and each file it changed: one `+` per inserted line.

`!git show --stat origin/main`

## Step 3 · Merge, then push again

`git merge` (Slide 19) joins their commit into yours in a new commit (Git's `ort` strategy); `--no-edit` keeps Git's message without opening an editor.

Run the cell. Before your edit, the merge prints `Already up to date.` because `main` is your own work. The last line starts with `PROBLEM: Git rejected your push`. On the TODO line, change `main` to `origin/main`:

`!git merge --no-edit origin/main`

Run the cell again. After your edit, the last line starts with `OK: your push was sent`.

```python
%step 3
!git merge --no-edit origin/main
!git push
```
