## The situation

A food delivery app's AI reads messages such as "My pizza came late" and replies with `order_id`, `issue`, `refund_amount` (None if not named) and `urgent`. Each of 8 made-up messages has a label: answers a person checked.

Run the cell below; it loads the data.

```python
# Given: messages, labels, prompts, simulated replies and scoring
FIELDS = ["order_id", "issue", "refund_amount", "urgent"]
MESSAGES = [  # the first three
    "Order A1042: my pizza came an hour late. Refund my $5 delivery fee.",
    "Order A1187: the soup spilled all over the bag. Send another.",
    "Order A1210 is someone else's dinner. Refund my $24 today.",
]
LABELS = [
    ("A1042", "late", 5, False), ("A1187", "damaged", None, False),
    ("A1210", "wrong_item", 24, True), ("A1305", "late", None, False),
    ("A1456", "wrong_item", None, True), ("A1501", "other", 8, False),
    ("A1533", "other", None, False), ("A1609", "damaged", 15, False),
]
# simulated: v1 writes 0 for None; v2 is right
REPLIES_V1 = [(o, i, 0 if r is None else r, u) for o, i, r, u in LABELS]
REPLIES_V2 = LABELS
PROMPT_V1 = """Reply with JSON only.
issue: late, damaged, wrong_item or other.
refund_amount: the money asked back.
If no amount is named, write 0 for refund_amount.
urgent: true if needed today."""
BAR = 0.9


def show_scores(replies):
    print("field          precision  recall  result")
    for i, field in enumerate(FIELDS):
        given = [r[i] == l[i] for r, l in zip(replies, LABELS) if r[i] is not None]
        p, r = sum(given) / len(given), sum(given) / sum(l[i] is not None for l in LABELS)
        verdict = "OK" if min(p, r) >= BAR else "PROBLEM: below the bar"
        print(f"{field:14} {p:9.2f} {r:7.2f}  {verdict}")


ANSWERS = {  # what each step should find, why, and where to look
    "SPOTTED_FIELD": ("refund_amount", "v1 wrote 0 where the label says None: no amount was named",
                      "Look at message 2: compare its label and its v1 reply."),
    "FAILING_FIELD": ("refund_amount", "its precision is 0.50, below the bar of 0.9",
                      "Find the row that says PROBLEM."),
    "CAUSE_LINE": ("4", "line 4 tells the AI to write 0 when no amount is named",
                   'Find the line that says "write 0".'),
    "KEEP": ("v2", "every field of v2 meets the bar", "Keep the prompt whose table has no PROBLEM."),
}


def review(name):
    value = globals().get(name)
    right, why, hint = ANSWERS[name]
    if value in ("", None):
        print(f"PROBLEM: {name} is still empty. Type your answer on the TODO line, then run this cell again.")
    elif str(value).strip().lower() == right:
        print(f"OK: {why}.")
    else:
        print(f"PROBLEM: {value!r} is not it. {hint}")
```

## Step 1 · Look before you measure

Run the cell: three messages, each with label and v1 reply. On the TODO line, type the field that differs; run again. Before you type, the last line starts with `PROBLEM: SPOTTED_FIELD is still empty.`; the right field gives a line that starts with `OK:`.

```python
# Step 1: show three messages
SPOTTED_FIELD = "refund_amount"

for n in range(3):
    print(f"Message {n + 1}: {MESSAGES[n]}")
    print("  label:   ", dict(zip(FIELDS, LABELS[n])))
    print("  v1 reply:", dict(zip(FIELDS, REPLIES_V1[n])))
review("SPOTTED_FIELD")
```

## Step 2 · Score prompt v1 against the bar

The bar: precision and recall both at least 0.9. Run the cell: one row per field. The last column says OK or PROBLEM. On the TODO line, type the field with the PROBLEM; run again. Before you type, the last line starts with `PROBLEM: FAILING_FIELD is still empty.`; the right field gives `OK:`.

```python
# Step 2: score the v1 replies
FAILING_FIELD = "refund_amount"

show_scores(REPLIES_V1)
review("FAILING_FIELD")
```

## Step 3 · Find the cause

A cause points at the fix (Slide 31).

Run the cell: prompt v1, numbered. On the TODO line, replace None with the number of the write 0 line; run again. Before you type, the last line starts with `PROBLEM: CAUSE_LINE is still empty.`; the right line gives `OK:`.

```python
# Step 3: show prompt v1 with its line numbers
CAUSE_LINE = 4

for number, line in enumerate(PROMPT_V1.splitlines(), start=1):
    print(number, line)
review("CAUSE_LINE")
```

## Step 4 · Score prompt v2 and choose

Prompt v2 says null, not 0, on line 4. Run the cell: v2's scores. On the TODO line, type `v1` or `v2`, whichever has no PROBLEM in its table; run again. Before you type, the last line starts with `PROBLEM: KEEP is still empty.`; the right prompt gives `OK:`.
