````python
# A copy of some of the challenge 1 starter's files, byte for byte
FILES = {
    "CHALLENGE.md": """# AI Full Stack Challenge — IT Access Requests

## The task
Northstar Labs receives access requests as free-text messages. Build a small runnable application that uses a language model to turn those messages into structured records and a review queue for the IT team. The supplied company, people and requests are fictional.

Read docs/requirements.md for the operating context and required behavior, materials/office-policy.md for review rules, and snapshot.schema.json for the record format. The 20 JSON exports in materials/requests/ are your starting data. Choose an interface suited to the IT team: a CLI, service or web application. Operators should be able to start intake, inspect completed results while other requests are pending, and resume unfinished work through the documented interface.

## What to submit
This brief ships as CHALLENGE.md; the README.md below is yours to write.

| Path | Content |
|---|---|
| README.md | Setup, supported runtime/provider, exact intake/test/accuracy-check commands, environment-variable names, and how to inspect the queue |
| Source code and dependency lockfile | Runnable application, extraction prompt, schema and model request configuration |
| deliverables/extractions.jsonl | One current record per message: document_id, source_sha256, extraction (schema-conforming object, or null on error), status, owner, next_action, error (null on success), and response_source (`mock`, or the provider and model ID that produced the record) |
| deliverables/queue.json | The review queue linked to extraction records by document_id |
| deliverables/accuracy.md | Source-derived expected fields for the 20 supplied messages and one paraphrase you author; the field-comparison procedure, exact-match counting/denominator rules, factual-error reporting and model/configuration; distinguish mocked examples from any optional live observations |
| deliverables/request-evidence.json | Credential-redacted example model request showing the model ID and configuration used, plus the implemented response handling |
| deliverables/verification.md | Verification commands and assertions, their intended comparisons, operating limits, and any observed results or repairs clearly identified as such |
| Coding-session capture | Session record paired with the submitted commit, collected through the facilitator's capture setup |

## Submission
Commit and push the application, outputs and session capture to your repository's default branch. Submit only the root repository link (`https://github.com/<your-account>/<repository>`) through the Classroom assignment. The facilitator supplies the assignment and capture instructions.

Submit the implemented provider integration and source-grounded validation procedure. Use mocked responses for reproducible development examples, and label each recorded observation with its configuration and whether it used a mock or provider response. The committed deliverables may come from mocked or provider responses; each record's response_source says which.
""",
    "README.md": """# AI Full Stack Challenge — IT Access Requests

The task is in CHALLENGE.md. Replace this file with your own README, as CHALLENGE.md asks.
""",
    "materials/requests/IT-001.json": """{
  "export_version": "1.0",
  "exported_at": "2026-10-22T16:00:00Z",
  "source_channel": "email",
  "subject": "Dashboard event checks",
  "message": "Maya Chen from Analytics here, with request IT-001 dated 2026-10-22. Can I get new read access on Atlas in staging from 2026-10-23 to 2026-11-06? My manager is Priya Shah. I'm checking the weekly conversion dashboard with the new event definition changes (before next reporting cycle). The sample records are already loaded, and I need to check the totals against our spreadsheet. Once I'm done with checking, I'll be sending out any discrepancies to the dashboard owner."
}
""",
}

from pathlib import Path

PRACTICE = Path("practice")
for name, text in FILES.items():
    path = PRACTICE / name
    if path.exists():
        print(f"Kept practice/{name}: it is already there.")
    else:
        path.parent.mkdir(parents=True, exist_ok=True)
        path.write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

## Check your work

Run the cell. Expect three `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: compare your three answers with what CHALLENGE.md says


def clean(answer: object) -> str:
    """Lower case, forward slashes, no quotes, no leading ./ or practice/, no trailing /."""
    text = str(answer).strip().strip("`'\"").replace("\\", "/").lower()
    return text.removeprefix("./").removeprefix("practice/").rstrip("/*")


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    if "TASK_FILE" not in globals():
        return "your answers are not saved. Run the Step 4 cell."
    if clean(TASK_FILE) != "challenge.md":
        return f"TASK_FILE is {TASK_FILE!r}. Which file holds the task? Ask Codex again (Step 3)."
    print("ok: the task is in CHALLENGE.md")
    if clean(INPUTS_FOLDER) != "materials/requests":
        return f"INPUTS_FOLDER is {INPUTS_FOLDER!r}. Which folder holds the 20 requests? Ask Codex again (Step 3)."
    print("ok: the inputs are in materials/requests/")
    if clean(HAND_IN).upper() != "B":
        return f"HAND_IN is {HAND_IN!r}. Read the choices in Step 4 again."
    print("ok: you hand in your own repository's link, through Classroom")
    return ""


problem = first_problem()
print(f"Not yet: {problem}" if problem else
      "Done: Codex explained the starter, and you know where challenge 1 starts and ends.")
````
