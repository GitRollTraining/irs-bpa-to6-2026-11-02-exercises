## Check your work

Run the cell. Expect three `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: read the hooks file, the commit messages and the checkpoint refs
import re
import subprocess
from pathlib import Path

PRACTICE = Path("practice")


def git(*args: str) -> str:
    try:
        run = subprocess.run(["git", "-C", str(PRACTICE), *args], capture_output=True, text=True)
    except OSError:
        return ""
    return run.stdout if run.returncode == 0 else ""


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    if not (PRACTICE / ".git").exists():
        return "practice/ has no Git repository. Do Step 2."
    if not (PRACTICE / ".codex" / "hooks.json").is_file():
        return "practice/.codex/hooks.json is missing. Do Step 3."
    print("ok: practice/.codex/hooks.json exists")
    linked = re.findall(r"^Entire-Checkpoint:[ \t]*(\S+)", git("log", "--format=%B"), re.M)
    if not linked:
        return "no commit has an Entire-Checkpoint: line. Redo Step 5, then Step 6."
    print(f"ok: a commit is linked to checkpoint {linked[0]}")
    refs = git("for-each-ref", "--format=%(refname)", "refs/entire/checkpoints/", "refs/heads/entire/checkpoints/v1")
    if not refs.strip():
        return "Git holds no checkpoint ref under refs/entire/checkpoints/. Run entire checkpoint list (Step 6)."
    print(f"ok: Git holds the checkpoint at {refs.split()[0]}")
    return ""


problem = first_problem()
print(f"Not yet: {problem}" if problem else
      "Done: your commit is linked to a recorded Codex session, and you checked it in Git.")
````
