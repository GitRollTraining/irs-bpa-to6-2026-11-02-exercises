````python
FILES = {
    "overdue.py": """from datetime import date


def overdue(cases: list[dict], today: date, days: int = 30) -> list[str]:
    \"\"\"Return the ids of open cases opened more than `days` days before today.

    Args:
        cases: Dicts with "id", "status" and "opened" (a date).
        today: The date to count from.
        days: The days a case may wait.

    Returns:
        The overdue ids.
    \"\"\"
    return [c["id"] for c in cases if c["status"] == "open" and (today - c["opened"]).days > days]
""",
    "broken/overdue_status.py": """# A broken copy, for Step 5.
from datetime import date


def overdue(cases: list[dict], today: date, days: int = 30) -> list[str]:
    \"\"\"Return the ids of open cases opened more than `days` days before today.

    Args:
        cases: Dicts with "id", "status" and "opened" (a date).
        today: The date to count from.
        days: The days a case may wait.

    Returns:
        The overdue ids.
    \"\"\"
    return [c["id"] for c in cases if (today - c["opened"]).days > days]
""",
    "broken/overdue_boundary.py": """# A broken copy, for Step 5.
from datetime import date


def overdue(cases: list[dict], today: date, days: int = 30) -> list[str]:
    \"\"\"Return the ids of open cases opened more than `days` days before today.

    Args:
        cases: Dicts with "id", "status" and "opened" (a date).
        today: The date to count from.
        days: The days a case may wait.

    Returns:
        The overdue ids.
    \"\"\"
    return [c["id"] for c in cases if c["status"] == "open" and (today - c["opened"]).days >= days]
""",
    ".gitignore": """__pycache__/
""",
    "test_overdue.py": """# Tests Codex wrote for overdue() (simulated), after Step 6.
from datetime import date

import pytest

from overdue import overdue

TODAY = date(2026, 11, 6)


@pytest.fixture
def cases() -> list[dict]:
    \"\"\"Build four synthetic cases.

    Returns:
        The cases C-1 to C-4.
    \"\"\"
    return [
        {"id": "C-1", "status": "open", "opened": date(2026, 9, 1)},
        {"id": "C-2", "status": "open", "opened": date(2026, 10, 31)},
        {"id": "C-3", "status": "closed", "opened": date(2026, 8, 1)},
        {"id": "C-4", "status": "open", "opened": date(2026, 10, 7)},
    ]


def test_only_c1_is_overdue(cases: list[dict]) -> None:
    assert overdue(cases, TODAY) == ["C-1"]


def test_closed_case_is_never_overdue(cases: list[dict]) -> None:
    assert "C-3" not in overdue(cases, TODAY)


def test_case_exactly_30_days_old_is_not_overdue(cases: list[dict]) -> None:
    assert "C-4" not in overdue(cases, TODAY)
""",
}

from pathlib import Path

PRACTICE = Path("practice")
for name, text in FILES.items():
    path = PRACTICE / name
    if path.exists():
        print(f"Kept practice/{name}: it is already there.")
    else:
        path.parent.mkdir(parents=True, exist_ok=True)
        path.write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")

# Codex's tests after Step 6 (simulated), committed after the start.
import subprocess

if not (PRACTICE / ".git").exists():
    subprocess.run(["git", "-C", str(PRACTICE), "init", "-q", "-b", "main"], check=True)
    subprocess.run(["git", "-C", str(PRACTICE), "add", "overdue.py", "broken", ".gitignore"], check=True)
    subprocess.run(["git", "-C", str(PRACTICE), "commit", "-qm", "Start"], check=True)
    subprocess.run(["git", "-C", str(PRACTICE), "add", "test_overdue.py"], check=True)
    subprocess.run(["git", "-C", str(PRACTICE), "commit", "-qm", "Add tests for overdue"], check=True)
    print("Committed the start and the tests")
````

### Learn: a fixture gives each test its own cases

A pytest fixture is a function that sets up data for tests. "At a basic level, test functions request fixtures they require by declaring them as arguments" ([pytest](https://docs.pytest.org/en/stable/how-to/fixtures.html)). Each test gets fresh cases, so no test changes another's.

## Check your work

Run the cell. It runs your tests on each copy of `overdue.py`. Expect three `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: run test_overdue.py on overdue.py and on each broken copy
import hashlib
import shutil
import subprocess
import sys
import tempfile
import xml.etree.ElementTree as ET
from pathlib import Path

PRACTICE = Path("practice")
START_SHA256 = "1dbe7acafe0390ae5b259eaaeb01fafd23268680c0e20d4a6111e771c7ad3107"


def failed_tests(broken=None):
    with tempfile.TemporaryDirectory() as tmp:
        work = shutil.copytree(PRACTICE, Path(tmp) / "p", ignore=shutil.ignore_patterns(".git", "__pycache__"))
        if broken:
            shutil.copyfile(work / "broken" / broken, work / "overdue.py")
        subprocess.run([sys.executable, "-m", "pytest", "-q", "-p", "no:cacheprovider", "test_overdue.py",
                        "--junitxml=r.xml"], cwd=work, capture_output=True, timeout=300)
        if not (work / "r.xml").is_file():
            return None
        return {case.get("name"): any(case.find(tag) is not None for tag in ("failure", "error", "skipped"))
                for case in ET.parse(work / "r.xml").iter("testcase")}


def first_problem():
    tests = PRACTICE / "test_overdue.py"
    if not tests.is_file():
        return "practice/test_overdue.py is missing. Give Codex the request (Step 4)."
    if "@pytest.fixture" not in tests.read_text(encoding="utf-8"):
        return "test_overdue.py has no @pytest.fixture. Ask Codex to put the cases in a fixture (Step 4)."
    print("ok: test_overdue.py builds the cases in a fixture")
    if hashlib.sha256((PRACTICE / "overdue.py").read_text(encoding="utf-8").encode()).hexdigest() != START_SHA256:
        return "overdue.py changed. The tests check the code: run git restore overdue.py (Step 5)."
    good = failed_tests()
    if not good:
        return "pytest found no test. Run python -m pytest -q in practice/ (Step 4)."
    failing = [name for name, failed in good.items() if failed]
    if failing:
        return f"{failing[0]} fails on overdue.py. Compare its expected value with Step 3."
    print(f"ok: all {len(good)} tests pass on the unchanged overdue.py")
    caught = set()
    for broken in ("overdue_status.py", "overdue_boundary.py"):
        found = [name for name, failed in (failed_tests(broken) or {}).items() if failed]
        if not found:
            return f"no test fails on broken/{broken}. Add a test that catches it (Step 6)."
        caught.update(found)
    never = [name for name in good if name not in caught]
    if never:
        return f"{never[0]} passes on both broken copies, so it can never fail. Fix or delete it (Step 6)."
    print("ok: each broken copy fails a test, and each test fails on a broken copy")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Run the Step 1 cell, then this cell."
print(f"Not yet: {problem}" if problem else "Done: Codex wrote the tests, you wrote the expected values, and each test can fail.")
````
