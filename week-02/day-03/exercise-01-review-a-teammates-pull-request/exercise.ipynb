### Step 2. Write the Code Review Rules, as your pull request

In the terminal, go to your clone and start a branch:

```sh
cd week-02/day-03/exercise-01-review-a-teammates-pull-request/lab
git switch -c review-rules
```

Open `lab/AGENTS.md` in the editor (in the Explorer: `week-02 > day-03 > exercise-01-review-a-teammates-pull-request > lab`). At its end, add a section `## Code Review Rules`. Write your own rules: at least four, numbered `1.`, `2.` and so on, one line each. Make each rule a thing a reviewer can check in a change. Cover these:

| A rule on | What it says | Why |
|---|---|---|
| Design | the change belongs in this code, and fits how the code works | "The most important thing to cover in a review is the overall design" |
| Functionality | the change does what its description says, edge cases included | "Does this CL do what the developer intended?" A CL is one change sent for review: read it as a pull request |
| Complexity | a reader understands it quickly, and it builds nothing the change does not need now | Over-engineering is code "where developers have made the code more generic than it needs to be, or added functionality that isn’t presently needed by the system." |
| Tests | each new rule or branch of the code has a test, and the tests pass | "Will the tests actually fail when the code is broken?" |

Number the rules: a review comment cites a rule by its number. Save the file. Then commit and push your branch:

```sh
git commit -am "Add Code Review Rules"
git push -u origin review-rules
```

On GitHub, you would now open a pull request from `review-rules` into `main`. Here, the pushed branch is your pull request. You should see `branch 'review-rules' set up to track 'origin/review-rules'`.

### Step 3. Make the change your teammate asks for, then squash-merge

Your teammate reads your pull request and asks for one change:

> "Add a rule on size, so a reviewer can ask for a split: a pull request holds one self-contained change of about 100 lines; a larger one is split."

Add that rule as the next numbered line of your section, in your own words, with the number 100 in it. Save the file. Commit and push the change to the same branch:

```sh
git commit -am "Add a rule on pull request size"
git push
```

Your reviewer approves your pull request. Squash-merge it into `main`. Then push `main`:

```sh
git switch main
git merge --squash review-rules
git commit -m "Add Code Review Rules"
git push
```

| Command | Why |
|---|---|
| `git merge --squash review-rules`, then `git commit` | "`git merge --squash` stages the branch's whole change and does not commit it." Git prints `Squash commit -- not updating HEAD`. Your commit records it as one commit, so `main` keeps one commit for each pull request. |
| `git push` | The team repository's `main` now holds your rules, so every branch made from it has them. |

Run cell 2. You should see one new commit on `main` and two on `review-rules`.

````python
# 2. See main and your branch. You do not need to read this code.
import subprocess
from pathlib import Path

lab = Path(_dh[0]) / "lab"
for branch in ("main", "review-rules"):
    log = subprocess.run(["git", "log", "--oneline", f"start..{branch}"], cwd=lab, capture_output=True, text=True)
    print(f"New on {branch}:", log.stdout.strip() or log.stderr.strip() or "nothing yet")
````
