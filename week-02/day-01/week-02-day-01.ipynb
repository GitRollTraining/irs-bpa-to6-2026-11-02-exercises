## 1.1 Protect main, keep branches short

*Who should be able to push to main, and who can change that rule?*

### Main stays known good; anyone can push until it is protected.

**Main**

- Fowler: the mainline, "a single, shared, branch"
- Microsoft: it should pass tests, build cleanly and always be current
- So every branch starts from a known good version

**By default**

- A personal repository: an owner and collaborators
- Collaborators can pull, and push (write)
- So every teammate can push to main

**Branches**

- A branch costs nothing to create
- What costs is the time it lives apart from main
- The filter: two branches, a few hours each

Martin Fowler calls it the mainline: "A single, shared, branch that acts as the current state of the product." Microsoft's guidance for its own teams says what that line must be: "The code in your main branch should pass tests, build cleanly, and always be current. Your main branch needs these qualities so that feature branches created by your team start from a known good version of code."

GitHub's permissions decide who can break it: "A repository owned by a personal account has two permission levels: the repository owner and collaborators." The second level can write: "Collaborators on a personal repository can pull (read) the contents of the repository and push (write) changes to the repository." So every teammate can push to main until someone protects it.

Microsoft: "Feature branches isolate work in progress from the completed work in the main branch. Git branches are inexpensive to create and maintain. Even small fixes and changes should have their own feature branch." Fowler: "Frequent integration increases the frequency of merges but reduces their complexity and risk. Frequent integration also alerts teams to conflicts much more quickly." The team's branches are status-filter-service and status-filter-page.

On the first morning, one person creates the team repository from the reference application, adds the others and protects main, including against themselves. The team then agrees three things in writing: how branches are named, who reviews whose pull requests, and how pull requests join main.

### Four settings the team turns on, and one default to change.

| Setting | What it does |
| --- | --- |
| Any protection rule | Disables force pushes; the branch cannot be deleted |
| Require pull request reviews | A push gets in only through an approved pull request |

| Require status checks | Every required check passes before a merge |

Any branch protection rule: "By default, each branch protection rule disables force pushes to the matching branches and prevents the matching branches from being deleted." Require pull request reviews: "If you enable required reviews, collaborators can only push changes to a protected branch via a pull request that is approved by the required number of reviewers with write permissions." Dismiss stale approvals: "Optionally, you can choose to dismiss stale pull request approvals when commits are pushed that affect the diff in the pull request." Require status checks: "After enabling required status checks, all required status checks must pass before collaborators can merge changes into the protected branch."

On a private repository protection needs a paid plan: "Protected branches are available in public repositories with GitHub Free and GitHub Free for organizations. Protected branches are also available in public and private repositories with GitHub Pro, GitHub Team, GitHub Enterprise Cloud, and GitHub Enterprise Server."

## 1.3 Merge, and resolve a conflict

*Theirs merged first, and both changed one function. What happens to yours?*

The question in full: you added the status filter to the function that builds the case list; a teammate changed the same function to sort newest first, and theirs merged first.

### Git merges three versions; the base shows both can stay.

<img src="figures/three-versions-one-resolution.png" width="864" alt="Left, grey, Base: return cases; the common ancestor. Arrows to two versions: blue, Yours (HEAD): if matches_status(...); filters by status. Berry, Theirs (main): sorted(..., reverse=True); sorts newest first; merged first. Both point into an orange box, The resolution: neither replaces the other: it filters, then sorts newest first. An arrow leads to Both sets of tests: run them, then git add marks it resolved.">

Git merges the two branch tips against their common ancestor; the base shows that neither change replaces the other, so the resolution keeps both, and both sets of tests run before git add.

Git combines three versions: "In this case, Git does a simple three-way merge, using the two snapshots pointed to by the branch tips and the common ancestor of the two." That works until both sides changed the same lines: "If you changed the same part of the same file differently in the two branches you're merging, Git won't be able to merge them cleanly." Update main and bring it into your branch with git merge main, and Git stops: "Git hasn't automatically created a new merge commit. It has paused the process while you resolve the conflict."

*Source: Pro Git, 2nd edition, 3.2 Basic Branching and Merging; git-merge manual; the chapter, section 1.5*

### With diff3, the conflict shows what both of you started from.

*The resolution: filter into matching, then sort matching newest first.*

| Between | The text | Whose |
| --- | --- | --- |
| <<<<<<< HEAD and \|\|\|\|\|\|\| | return [case for case in cases if matches_status(case, status)] | Typically your side |
| \|\|\|\|\|\|\| base and ======= | return cases | The original text, which diff3 adds |
| ======= and >>>>>>> main | return sorted(cases, key=lambda case: case.opened_on, reverse=True) | Typically their side |

### The resolution is code neither of you tested.

**Before git add**

- Run both teammates' tests
- Then git add each file: staging marks it resolved

**An agent's resolution**

- A coding agent can propose one
- Only the base and both sets of tests show it kept what both meant

**Merge, not rebase**

- Do not rebase commits others may have built on
- Once a branch is on the team repository, bring main in with a merge

Then: "After you've resolved each of these sections in each conflicted file, run git add on each file to mark it as resolved. Staging the file marks it as resolved in Git." A coding agent can propose a resolution; only the base version and both sets of tests show it kept what both sides meant. One rule for how you bring main in, from Pro Git: "Do not rebase commits that exist outside your repository and that people may have based work on." Once a branch is on the team repository, bring main in with a merge.

## 1.4 Python for Java, C# or COBOL

*Which lines of a Python status filter would a Java or C# reader misread?*

### Four things a Java or C# reader can miss in Python.

| In Python | What happens | What to do |
| --- | --- | --- |
| Dynamic semantics | Many errors appear only when a line runs | Tests carry the checks a compiler made |
| Names, by PEP 8 | Functions and variables lowercase, with underscores | Review against PEP 8 |
| except Exception: return [] | Every fault becomes an empty case list | Catch only the exceptions you expect |
| A mutable default | Created once, shared by every call | Default to None; make the list inside |

The Python Software Foundation's own description: "Python is an interpreted, object-oriented, high-level programming language with dynamic semantics." One consequence is that many errors appear only when a line runs: "Even if a statement or expression is syntactically correct, it may cause an error when an attempt is made to execute it." A misspelt attribute in a branch no test runs is not an error until a user reaches it, so in Python the tests carry checks a compiler made for you in C# or Java.

You will review against PEP 8, the Python project's style guide, whose premise is that "code is read much more often than it is written." Where it differs most from a Java or C# shop: "Function names should be lowercase, with words separated by underscores as necessary to improve readability. Variable names follow the same convention as function names."

The tutorial: "However, it is good practice to be as specific as possible with the types of exceptions that we intend to handle, and to allow any unexpected exceptions to propagate on." A generated handler that wraps the request in except Exception: return [] turns every fault in the filter into an empty case list, the hardest fault to see.

### A default is made once, so every call shares one list.

<img src="figures/one-default-list-shared-by-every-call.png" width="864" alt="Left, the code: def active_filters(params, filters=[]): if &#x27;status&#x27; in params: filters.append((&#x27;status&#x27;, params[&#x27;status&#x27;])) return filters. filters=[] is ringed in orange: filters=[] is made once, when the function is defined; a dashed orange line runs from it to an orange box on the right, One list, shared: every call gets this same list. Above that box, Request 1: status open: appends (&#x27;status&#x27;, &#x27;open&#x27;), with an arrow into the list. Below it, Request 2: no status: still filters by open, with an arrow from the list. Under the code, in blue, The fix: default to None: check for None inside the function and create a new list. The status open is an example value.">

The FAQ's fix: "Because of this feature, it is good programming practice to not use mutable objects as default values. Instead, use None as the default value and inside the function, check if the parameter is None and create a new list/dictionary/whatever if it is."

*Source: Python documentation, Programming FAQ; the chapter, section 2.1*

Mapping: in a study of 32 people extending unfamiliar Python code with the authors' tool, GILT, or with web search, "professionals completed 0.57 more sub-tasks with GILT support compared to when they used search engines, whereas students did not experience significant gains." A professional there is anyone who was not a student, and 13 of those 16 were researchers. The authors' models found no significant difference in completion time or quiz scores, which they qualify by the statistical power of the experiment.
