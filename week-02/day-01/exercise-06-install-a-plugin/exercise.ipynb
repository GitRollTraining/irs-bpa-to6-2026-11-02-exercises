# Exercise 6 · Install a ready-made skills plugin

You will add a plugin to Codex, run its setup skill in a small project, and check the folder it makes.

**Goal:** Install the apprentice workspace plugin in Codex, then run its `setup` skill in a practice project.

**Why it matters:** One skill does one job. A plugin brings many skills, rules and templates in one install.

**Before you start:**
- Codex is installed and signed in, and your computer is online.
- A skill is a folder with a `SKILL.md` file of instructions for one kind of task.

**What the check measures:** `practice/apprentice-workspace/AGENTS.md` exists, and `practice/AGENTS.md` holds the plugin's block.

**Done when:** the last cell prints `Done:`.

## The situation

You want the same folders and rules in every new project. A plugin can set them up for you. The plugin comes from another GitRoll course, AI Fellowship, so its files use that name. The project files are made up.

## Step 1 · Write the practice files

Run the cell. It writes `README.md` and `.gitignore` into `practice/`. It never overwrites a file.

Expect two lines that start with `Wrote practice/`, or `Kept` for a file that is already there.

````python
FILES = {
    "README.md": """# Practice project

A small project for trying a skills plugin.
""",
    ".gitignore": "__pycache__/\n",
}

from pathlib import Path

PRACTICE = Path("practice")
PRACTICE.mkdir(exist_ok=True)
for name, text in FILES.items():
    if (PRACTICE / name).exists():
        print(f"Kept practice/{name}: it is already there.")
    else:
        (PRACTICE / name).write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

## Start over (only if you need it)

Set `START_OVER = True` and run the cell. It writes the Step 1 files again. Git keeps your commits. Then set it back to `False` and go to Step 4.

Expect `Nothing changed.` while it is `False`.

````python
# Start over: only if START_OVER is True, write the Step 1 files again
START_OVER = False

if not START_OVER:
    print("Nothing changed.")
elif "FILES" not in globals():
    print("Run the Step 1 cell first, then this cell.")
else:
    PRACTICE.mkdir(exist_ok=True)
    for name, text in FILES.items():
        (PRACTICE / name).write_text(text, encoding="utf-8")
    print("Wrote the practice files again. Now set START_OVER back to False.")
````

## Check your work

Run the cell. Expect two `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: find the files the setup skill makes
from pathlib import Path

PRACTICE = Path("practice")
MARKER = "<!-- apprentice-workspace: start -->"


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    if not PRACTICE.is_dir():
        return "practice/ is missing. Run the Step 1 cell."
    if not (PRACTICE / "apprentice-workspace" / "AGENTS.md").is_file():
        return "practice/apprentice-workspace/AGENTS.md is missing. Install the plugin (Step 3), then run setup (Step 4)."
    print("ok: practice/apprentice-workspace/AGENTS.md exists")
    root = PRACTICE / "AGENTS.md"
    if not root.is_file() or MARKER not in root.read_text(encoding="utf-8"):
        return "practice/AGENTS.md has no apprentice-workspace block. Run setup again (Step 4)."
    print("ok: practice/AGENTS.md tells Codex to read the workspace rules")
    return ""


problem = first_problem()
print(f"Not yet: {problem}" if problem else
      "Done: one install gave Codex a set of skills, and one skill set up your project.")
````
