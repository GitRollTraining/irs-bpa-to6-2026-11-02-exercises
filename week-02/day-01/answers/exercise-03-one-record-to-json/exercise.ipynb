## Step 1 · Write the practice files

This is the solved copy. Run the cell. It writes the two Step 1 files and a correct `record.json`, written by hand from the message. It never overwrites a file.

Expect three lines that start with `Wrote practice/`.

````python
FILES = {
    "snapshot.schema.json": """{
  "type": "object",
  "properties": {
    "request_id": {
      "type": [
        "string",
        "null"
      ]
    },
    "request_date": {
      "type": [
        "string",
        "null"
      ]
    },
    "kind": {
      "type": [
        "string",
        "null"
      ],
      "enum": [
        "new",
        "change",
        "withdrawal",
        "other",
        null
      ]
    },
    "requester": {
      "type": [
        "string",
        "null"
      ]
    },
    "team": {
      "type": [
        "string",
        "null"
      ]
    },
    "system": {
      "type": [
        "string",
        "null"
      ]
    },
    "environment": {
      "type": [
        "string",
        "null"
      ],
      "enum": [
        "staging",
        "production",
        null
      ]
    },
    "permission": {
      "type": [
        "string",
        "null"
      ],
      "enum": [
        "read",
        "write",
        "admin",
        null
      ]
    },
    "business_purpose": {
      "type": [
        "string",
        "null"
      ]
    },
    "start_date": {
      "type": [
        "string",
        "null"
      ]
    },
    "end_date": {
      "type": [
        "string",
        "null"
      ]
    },
    "manager": {
      "type": [
        "string",
        "null"
      ]
    },
    "supporting_quotes": {
      "type": "array",
      "items": {
        "type": "string"
      }
    },
    "issues": {
      "type": "array",
      "items": {
        "type": "string"
      }
    }
  },
  "required": [
    "request_id",
    "request_date",
    "kind",
    "requester",
    "team",
    "system",
    "environment",
    "permission",
    "business_purpose",
    "start_date",
    "end_date",
    "manager",
    "supporting_quotes",
    "issues"
  ],
  "additionalProperties": false
}
""",
    "materials/requests/IT-001.json": """{
  "export_version": "1.0",
  "exported_at": "2026-10-22T16:00:00Z",
  "source_channel": "email",
  "subject": "Dashboard event checks",
  "message": "Maya Chen from Analytics here, with request IT-001 dated 2026-10-22. Can I get new read access on Atlas in staging from 2026-10-23 to 2026-11-06? My manager is Priya Shah. I'm checking the weekly conversion dashboard with the new event definition changes (before next reporting cycle). The sample records are already loaded, and I need to check the totals against our spreadsheet. Once I'm done with checking, I'll be sending out any discrepancies to the dashboard owner."
}
""",

    # A correct record for IT-001, written by hand from the message (simulated Codex output)
    "record.json": """{
  "request_id": "IT-001",
  "request_date": "2026-10-22",
  "kind": "new",
  "requester": "Maya Chen",
  "team": "Analytics",
  "system": "Atlas",
  "environment": "staging",
  "permission": "read",
  "business_purpose": "Check the weekly conversion dashboard with the new event definition changes before the next reporting cycle.",
  "start_date": "2026-10-23",
  "end_date": "2026-11-06",
  "manager": "Priya Shah",
  "supporting_quotes": [
    "Maya Chen from Analytics here, with request IT-001 dated 2026-10-22.",
    "Can I get new read access on Atlas in staging from 2026-10-23 to 2026-11-06?",
    "My manager is Priya Shah.",
    "I'm checking the weekly conversion dashboard with the new event definition changes (before next reporting cycle)."
  ],
  "issues": []
}
""",
}

from pathlib import Path

PRACTICE = Path("practice")
for name, text in FILES.items():
    path = PRACTICE / name
    if path.exists():
        print(f"Kept practice/{name}: it is already there.")
    else:
        path.parent.mkdir(parents=True, exist_ok=True)
        path.write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

## Check your work

Run the cell. Expect five `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: compare the inputs, read record.json, validate it, and compare it with the message
import hashlib
import json
from pathlib import Path

PRACTICE = Path("practice")
INPUTS = {
    "snapshot.schema.json": "2802bed75be35046552b6d23f45447b416b440fc8ba5a62a9d1d50d0eebbc636",
    "materials/requests/IT-001.json": "102a69f8fbc3779ec19fb6c761ad34b0c9a001f845b8df3c808f5a79441fbb10",
}
FROM_MESSAGE = {"request_id": "IT-001", "environment": "staging", "permission": "read"}


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    for name, digest in INPUTS.items():
        path = PRACTICE / name
        if not path.is_file() or hashlib.sha256(path.read_bytes()).hexdigest() != digest:
            return f"practice/{name} is missing or changed. Write it again with Start over, then go to Step 3."
    print("ok: the schema and the request are unchanged")
    try:
        record = json.loads((PRACTICE / "record.json").read_text(encoding="utf-8"))
    except FileNotFoundError:
        return "practice/record.json is missing. Ask Codex for it (Step 3)."
    except ValueError as error:
        return f"practice/record.json is not valid JSON ({error}). Ask Codex to fix it (Step 3)."
    print("ok: record.json is valid JSON")
    try:
        from jsonschema import Draft202012Validator, exceptions
    except ImportError:
        return "the jsonschema package is missing. Run uv sync in the course folder (course README)."
    schema = json.loads((PRACTICE / "snapshot.schema.json").read_text(encoding="utf-8"))
    error = exceptions.best_match(Draft202012Validator(schema).iter_errors(record))
    if error:
        where = "/".join(str(part) for part in error.absolute_path) or "the top"
        return f"record.json does not fit the schema at {where}: {error.message}. Ask Codex to fix it (Step 3)."
    print("ok: snapshot.schema.json accepts record.json")
    for field, value in FROM_MESSAGE.items():
        if record[field] != value:
            return f"record.json has {field} = {record[field]!r}, but the message says {value!r}. See Step 4."
    print("ok: request_id, environment and permission match the message")
    message = json.loads((PRACTICE / "materials/requests/IT-001.json").read_text(encoding="utf-8"))["message"]
    if not record["supporting_quotes"]:
        return "record.json has no supporting quotes. Ask Codex to copy them from the message (Step 3)."
    for quote in record["supporting_quotes"]:
        if quote not in message:
            return f"this quote is not word for word in the message: {quote!r}. Ask Codex to copy it exactly (Step 3)."
    print("ok: every supporting quote is word for word in the message")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Step 1 writes back a missing file."
print(f"Not yet: {problem}" if problem else "Done: one request is now one JSON record that the challenge's schema accepts.")
````
