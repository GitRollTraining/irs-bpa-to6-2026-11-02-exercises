### Step 5. Check

Run cell 4. It prints `Done:` when four things are true:

- `AGENTS.md` names `pytest`.
- `AGENTS.md` names `flask --app app run`.
- Codex changed a file of the app, and the tests ran after its last change and after `AGENTS.md`.
- The app's tests pass.

Otherwise it prints `Not yet:` and each thing that is still missing. You do not need to read its code.

````python
# 4. Check. You do not need to read this code.
%cd -q {_dh[0]}/reference-app
import subprocess
import sys
from pathlib import Path


def newest(paths: list[Path]) -> float:
    """Return the newest modification time of these files, or 0.0 when there are none."""
    return max((p.stat().st_mtime for p in paths if p.is_file()), default=0.0)


def check() -> list[str]:
    """Return what is left: one line for each thing that is not yet right."""
    left = []
    agents = Path("AGENTS.md")
    text = agents.read_text(encoding="utf-8") if agents.exists() else ""
    if "pytest" not in text:
        left.append("AGENTS.md does not name pytest. Type step 2's tests line in cell 2.")
    if "flask --app app run" not in text:
        left.append("AGENTS.md does not name flask --app app run. Type step 2's app line in cell 2.")
    changed = subprocess.run(["git", "status", "--porcelain", "--untracked-files=all", "--", "."],
                             capture_output=True, text=True).stdout.splitlines()
    app_files = [Path(line[3:]).name for line in changed
                 if not line[3:].endswith("AGENTS.md") and "/.pytest_cache/" not in line
                 and "__pycache__" not in line]
    test_run = newest(list(Path(".pytest_cache").rglob("*")))
    last_change = newest([p for p in Path(".").rglob("*") if p.suffix in (".py", ".html", ".json")
                          and ".pytest_cache" not in p.parts and "__pycache__" not in p.parts])
    if not app_files:
        left.append("Codex has not changed a file of the app yet. Ask for the change in step 3.")
    elif not agents.exists() or test_run <= max(last_change, agents.stat().st_mtime):
        left.append("The tests did not run after Codex's last change and after AGENTS.md. Check "
                    "that AGENTS.md names pytest. Start a new Codex session, which reads AGENTS.md, "
                    "and ask for the change again.")
    tests = subprocess.run([sys.executable, "-m", "pytest", "-q", "-p", "no:cacheprovider", "--color=no"],
                           capture_output=True, text=True)
    if "No module named pytest" in tests.stderr:
        left.append("The kernel is not week 2's environment. Select it as step 1 says.")
    elif tests.returncode:
        last = (tests.stdout.strip() or tests.stderr.strip()).splitlines()[-1]
        left.append(f"The app's tests do not pass ({last}). Ask Codex to fix what it changed.")
    return left


left = check()
print(f"Not yet: {len(left)} left." if left else "Done: AGENTS.md names pytest and the app's "
      "command, Codex ran the tests after its change, and they pass.")
for line in left:
    print("- " + line)
````
