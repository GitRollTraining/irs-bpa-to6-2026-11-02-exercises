## The situation

`reading_time.py` tells how many minutes a text takes to read, at 200 words a minute. You want the reading speed to be a choice. The files are made up.

## Step 1 · Write the practice files

This is the solved copy. Run the cell. It writes `practice/` as it is after Step 5: the `Start` commit, then your `AGENTS.md`, Codex's change and the `CHANGES.md` it wrote.

Expect `Committed the start`, then four `Wrote practice/` lines.

````python
FILES = {
    "reading_time.py": """\"\"\"Estimate how long a text takes to read.\"\"\"
import math


def minutes_to_read(text: str) -> int:
    \"\"\"Return the minutes needed to read text at 200 words a minute, at least 1.\"\"\"
    return max(1, math.ceil(len(text.split()) / 200))
""",
    "test_reading_time.py": """from reading_time import minutes_to_read


def test_minutes_to_read():
    assert minutes_to_read("word " * 450) == 3
""",
    "pytest.ini": "[pytest]\n",
    ".gitignore": "__pycache__/\n.pytest_cache/\n",
}

# The files after Step 5 (simulated): your AGENTS.md, Codex's change, and the CHANGES.md it wrote
AFTER = {
    "reading_time.py": """\"\"\"Estimate how long a text takes to read.\"\"\"
import math


def minutes_to_read(text: str, words_per_minute: int = 200) -> int:
    \"\"\"Return the minutes needed to read text at words_per_minute, at least 1.\"\"\"
    return max(1, math.ceil(len(text.split()) / words_per_minute))
""",
    "test_reading_time.py": """from reading_time import minutes_to_read


def test_minutes_to_read():
    assert minutes_to_read("word " * 450) == 3


def test_minutes_to_read_words_per_minute():
    assert minutes_to_read("word " * 450, words_per_minute=100) == 5
""",
    "AGENTS.md": """# Rules for this project
- After every change, run the tests with python -m pytest -q.
- After every change, add one line to CHANGES.md that names the file you changed.
""",
    "CHANGES.md": """- reading_time.py: minutes_to_read takes words_per_minute, default 200.
- test_reading_time.py: a test for words_per_minute.
""",
}

import subprocess
from pathlib import Path

PRACTICE = Path("practice")
if (PRACTICE / ".git").exists():
    print("Kept practice/: it is already there.")
else:
    PRACTICE.mkdir(exist_ok=True)
    for name, text in FILES.items():
        (PRACTICE / name).write_text(text, encoding="utf-8")
    subprocess.run(["git", "-C", str(PRACTICE), "init", "-q", "-b", "main"], check=True)
    subprocess.run(["git", "-C", str(PRACTICE), "add", "-A"], check=True)
    subprocess.run(["git", "-C", str(PRACTICE), "commit", "-qm", "Start"], check=True)
    print("Committed the start")
    for name, text in AFTER.items():
        (PRACTICE / name).write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

## Check your work

Run the cell. Expect three `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: read AGENTS.md and CHANGES.md, ask Git what changed, and run the tests
import contextlib
import io
import re
import subprocess
import sys
from pathlib import Path

import pytest

PRACTICE = Path("practice")


def git(*args: str) -> list:
    """Run git in practice/ and return the words it prints."""
    return subprocess.run(["git", "-C", str(PRACTICE), *args], capture_output=True, text=True).stdout.split()


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    rules = PRACTICE / "AGENTS.md"
    if not rules.is_file() or "CHANGES.md" not in rules.read_text(encoding="utf-8"):
        return "practice/AGENTS.md is missing or does not name CHANGES.md. Write your rule (Step 3)."
    print("ok: AGENTS.md names CHANGES.md")
    if not (PRACTICE / ".git").exists():
        return "practice/ has no Git repository. Do Step 2."
    start = git("rev-list", "--max-parents=0", "HEAD")
    if not start:
        return "practice/ has no Start commit. Commit the start (Step 2)."
    changed = set(git("diff", "--name-only", start[-1]) + git("ls-files", "--others", "--exclude-standard"))
    changed -= {"AGENTS.md", "CHANGES.md"}
    notes = PRACTICE / "CHANGES.md"
    text = notes.read_text(encoding="utf-8") if notes.is_file() else ""
    named = [n for n in sorted(changed) if re.search(rf"(?<![\w-]){re.escape(n)}(?![\w-])", text)]
    if not named:
        return ("practice/CHANGES.md is missing or names no file that changed since Start. "
                "Give Codex the task (Step 4), then see Step 5.")
    print(f"ok: CHANGES.md names {', '.join(named)}")
    for name in ("reading_time", "test_reading_time"):  # so pytest reads the files again
        sys.modules.pop(name, None)
    with contextlib.redirect_stdout(io.StringIO()) as out, contextlib.redirect_stderr(io.StringIO()):
        exit_code = pytest.main(["-q", "--color=no", "-p", "no:cacheprovider", str(PRACTICE)])
    if exit_code != 0:
        return "the tests fail. Run python -m pytest -q in practice/ to see why, and ask Codex to fix it."
    print("ok: the tests pass")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Step 1 writes back a missing file."
print(f"Not yet: {problem}" if problem else "Done: your AGENTS.md rule left CHANGES.md, naming a file Codex changed.")
````
