## Step 1 · Write the practice files

Run the cell. It writes `dates.py` and `.gitignore` into `practice/`, beside this notebook. It never overwrites a file.

Expect two lines that start with `Wrote practice/`, or `Kept` for a file that is already there.

````python
FILES = {
    "dates.py": """\"\"\"Small helpers for dates written as YYYY-MM-DD.\"\"\"
from datetime import date


def today_text() -> str:
    \"\"\"Return today's date as YYYY-MM-DD.\"\"\"
    return date.today().isoformat()
""",
    ".gitignore": "__pycache__/\n",
}

from pathlib import Path

PRACTICE = Path("practice")
for name, text in FILES.items():
    path = PRACTICE / name
    if path.exists():
        print(f"Kept practice/{name}: it is already there.")
    else:
        path.parent.mkdir(parents=True, exist_ok=True)
        path.write_text(text, encoding="utf-8")
        print(f"Wrote practice/{name}")
````

## Check your work

Run the cell. Expect three `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

````python
# Check your work: find the skill's SKILL.md, read its front matter, and compare your answer
import re
from pathlib import Path

PRACTICE = Path("practice")


def front_matter(skill_md: Path) -> str:
    """Return the text between the first two --- lines of SKILL.md, or ""."""
    parts = skill_md.read_text(encoding="utf-8").split("---", 2)
    return parts[1] if len(parts) == 3 and not parts[0].strip() else ""


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    found = sorted((PRACTICE / ".agents" / "skills").glob("*/SKILL.md"))
    if not found:
        return ("practice/.agents/skills/ holds no skill folder with a SKILL.md. Install from practice/ "
                "and choose the scope Project, not Global (Step 5).")
    print(f"ok: found {', '.join(path.parent.name for path in found)} in practice/.agents/skills/")
    for skill_md in found:
        head = front_matter(skill_md)
        if re.search(r"^name:\s*\S", head, re.M) and re.search(r"^description:\s*\S", head, re.M):
            print(f"ok: {skill_md.parent.name}/SKILL.md has a name and a description")
            break
    else:
        return "no SKILL.md in practice/.agents/skills/ has a name and a description. Install the skill again (Step 5)."
    choice = str(globals().get("ANSWER", "")).strip().upper()
    if not choice:
        return "the question has no answer. Write a letter in the Step 7 cell and run it."
    if choice != "A":
        return f"answer {choice} is not right. Read the Learn cell before Step 7, then try again."
    print("ok: your answer is right")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Install the skill again (Step 5)."
print(f"Not yet: {problem}" if problem else "Done: a skill is installed in practice/, and you know what it is.")
````
