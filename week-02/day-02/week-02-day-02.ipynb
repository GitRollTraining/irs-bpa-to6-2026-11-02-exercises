## 2.1 Measure the coding agent

*Your team built the status filter with the coding agent. Did it save you time, and what would show it?*

### Developers felt faster; measured, they took 19% longer.

*How the work felt is not evidence.*

- 16 developers, 246 tasks, their own repositories
- Each task: AI allowed or forbidden, at random

<img src="figures/forecast-faster-measured-slower.png" width="864" alt="Three bars on one scale around a line marked no change: less time with AI to the left, more time with AI to the right. Forecast, before the tasks: 24% less, in grey. Estimate, after the study: 20% less, in grey. Measured: 19% more, in orange.">

METR ran a randomized trial with 16 experienced open-source developers on 246 tasks in their own large repositories, each task assigned at random to allow AI tools or forbid them.

The authors limit their own result: the slowdown does not mean the tools rarely help, and "we find evidence that the high developer familiarity with repositories and the size and maturity of the repositories both contribute to the observed slowdown, and these factors do not apply in many software development settings."

*Source: METR, Measuring the Impact of Early-2025 AI on Experienced Open-Source Developer Productivity; the chapter, section 2.7*

### Other trials: the opposite sign, each measuring something else.

| Study | Who, and what work | Result, against what |
| --- | --- | --- |
| Paradis and others | 96 Google engineers, one complex task | About 21% less time, a wide interval |
| Peng and others | 95 programmers, an HTTP server in JavaScript | 55.8% less time than a group without Copilot |
| Cui, Demirer and others | Developers at three companies, normal work | 26.08% more pull requests a week |

Other trials in other settings found the opposite sign, each comparing a group with the tools against a group without them, and each measuring something different. Paradis and others: 96 Google engineers on one complex task with Google's own AI features, about 21% less time on task than without them, with a wide interval. Peng and others: 95 recruited programmers writing an HTTP server in JavaScript, 55.8% less time than a control group without Copilot. Cui, Demirer and others: developers at three companies in their normal work, 26.08% more pull requests a week with the tool than without it, pooled across the three.

Peng's study says what it left out: "this study does not examine the effects of AI on code quality." Someone else's number does not tell you what the agent does on your repository; your own measurement does.

### Pair the runs; record time, defects and review effort.

*Each run records four rows:*

- Time, start to merge
- Defects, and where found
- Review effort
- What the tool did

<img src="figures/each-teammate-runs-it-both-ways.png" width="864" alt="Two columns: Task 1, the tax-year filter; Task 2, another of the same size. Each run records four rows: time from start to merge, review and fixes included; defects, and where each was found; review effort in rounds, comments and reviewer minutes; and what the tool did, and what you checked.

The SPACE framework, from researchers at GitHub and Microsoft, holds that productivity "cannot be measured by a single metric or activity data alone", and advises: "To measure developer productivity, teams and leaders (and even individuals) should capture several metrics across multiple dimensions of the framework—at least three are recommended." The measured exercise records three, chosen by this course to sit in three of SPACE's dimensions: time from start to merge (efficiency and flow), defects with where each was found (performance), and review effort in rounds, comments and reviewer minutes (communication and collaboration). The fourth row, what the tool did and what you checked, is left empty for the run without it.

Two teammates build it, one with the coding agent and one without, then swap on another task of the same size, so each of you has one run of each and whoever drew the easier task does not decide the result.

*Source: the chapter, section 2.7, and The SPACE of Developer Productivity*

## 2.2 Read the diff, not the description

*The description is true and every test passes. What is left to read?*

The question in full: a generated pull request says it fixes the status filter and that all tests pass, and both statements are true.

### A description is part of the change: read tests and CI first.

*Any change that weakens CI is a blocker. B touches the filter and its test file.*

<img src="figures/read-the-tests-before-the-description.png" width="864" alt="Four steps in a row, joined by arrows: CI and test files, workflows, tests, coverage, build scripts, in orange; The rest of the diff, every name it calls, opened; One sentence, what it does, from the diff alone; Description, last, compared with your sentence. The first step opens onto pull request B&#x27;s test file, read first: def test_filter_in_review():, a removed line cases = [make_case(status=&#x27;In Review&#x27;)], an added line, ringed in orange, cases = [make_case(status=&#x27;in_review&#x27;)], and assert len(filter_cases(cases, &#x27;In Review&#x27;)) == 1. An orange arrow from the added line points to: Never stored: &#x27;in_review&#x27;; the application stores &#x27;In Review&#x27;, so the test passes against code that would still return nothing for a real In Review case.">

Simon Willison: "Agents write convincing looking pull request descriptions. You need to review these too! It's rude to expect someone else to read text that you haven't read and validated yourself."

Then read the rest, opening every name the diff calls, write one sentence saying what the change does, from the diff alone, and only then compare it with the description.

The test now builds a case with a status the application never stores, so it passes against code that would still return nothing for a real In Review case.

*Source: GitHub Blog, Agent pull requests are everywhere. Here's how to review them.; Simon Willison; the chapter, section 3.1*

### A runs and returns cases: four ways it can still be wrong.

| Pattern, as the study names it | In pull request A |
| --- | --- |
| Misinterpretation: most common, 20.77% | Filters on the year the case was opened, not the tax year |
| Missing Corner Case: second, 15.27% | No test for no year, a year with no cases, or a year as text |
| Hallucinated Object | Calls a helper defined nowhere: open every name it calls |
| Non-Prompted Consideration | Also re-sorts the list, which nobody asked for |

One catalogue comes from a study that classified 333 bugs in code from three models: CodeGen, PanGu-Coder and Codex, an earlier OpenAI model, not today's Codex CLI. Its warning to users: "Our analysis and the feedback received from survey participants also reveal that LLM-generated code can appear correct at first glance and deceive users depending on the complexity of the code and the experience of said users." Four of the ten patterns it names, the first two its most common: Misinterpretation, "among the most common in our analyzed dataset across all three models, with 20.77% of buggy code samples"; Missing Corner Case, "the second most common pattern with 15.27%"; Hallucinated Object, "when the LLM hallucinates by attempting to utilize an object that neither exists nor has been defined"; Non-Prompted Consideration, "when the LLM adds statements to the code that are unrelated to the task specified in the prompt, leading to errors."

A study of patches that three issue-solving tools produced for the SWE-bench Verified benchmark, each passing the benchmark's tests, found that "even more (29.6%) plausible patches induce different behavior than the ground truth patches." Behaving differently from what a developer wrote is not always wrong; it is what you read the diff for.

### Pull request C: the status reaches the SQL. Parameterize it.

*Then check the new package and any secret.*

- 19.7% of generated packages are fictitious
- A credential in the change: revoke or rotate it

<img src="figures/the-status-reaches-the-query.png" width="864" alt="Left, The status, from the request, with two arrows to the right. The upper arrow, orange, goes to Pull request C: pasted into the SQL text: the database may run part of it as a command. The lower arrow, blue, goes to Parameterized: passed apart from the SQL text: WHERE status = ?">

Check first where the status from the request goes. In this illustration the service reads cases from a database, and the diff replaces the filter over load_cases() with: sql = '''SELECT id, tax_year, status FROM cases WHERE status = '%s' ''' % status, then db.execute(sql).fetchall().

OWASP, the open project that ranks web application risks, defines what that line allows: "An injection vulnerability is an application flaw that allows untrusted user input to be sent to an interpreter (e.g. a browser, database, the command line) and causes the interpreter to execute parts of that input as commands." SQL injection is second, after cross-site scripting, in the 2025 list of the most dangerous software weaknesses, ranked from 39,080 published vulnerability records. OWASP's remedy: "The preferred option is to use a safe API, which avoids using the interpreter entirely, provides a parameterized interface, or migrates to Object Relational Mapping Tools (ORMs)." The fixed lines: sql = 'SELECT id, tax_year, status FROM cases WHERE status = ?', then db.execute(sql, (status,)).fetchall().

Do not expect the author's confidence to warn you. In a study of 47 participants, mostly university students, writing security-related code with and without an AI code generator, those with access wrote significantly less secure code, and were more likely to believe they had written secure code.

A study of 576,000 code samples from 16 models found that "19.7% of the generated packages are fictitious, posing a critical threat to software security through package confusion attacks." The risk is that someone registers the name, and checking that a name exists does not help. A credential in a generated change is a leak like any other, and GitHub's first step is to "revoke and/or rotate that secret." A fixture holding real taxpayer data breaks week 1's data rules before anything else.

*Source: OWASP Top 10:2025, A05 Injection; the chapter, section 3.4*

### Most rejections give no reason; a refusal names three things.

<img src="figures/a-refusal-in-three-parts.png" width="864" alt="A review comment, Request changes., in three blocks, each joined to its name at the left. In blue, What you found, with the line: Found: tests/test_case_list.py, test_filter_in_review now builds a case with status &#x27;in_review&#x27;. The application stores &#x27;In Review&#x27;, so the test no longer tests a real case. In berry, Why it blocks the merge: Why it blocks: with the original test restored, the fix returns no cases for In Review, which is the fault it says it fixes. In orange, What would make it acceptable: What would make it acceptable: the original test unchanged and passing; a new test that fails on main and passes here for a stored status with a space; and a note saying how statuses are compared, so the page and the service agree.">

A refusal that carries no reason teaches nobody anything, and most rejections carry none. A study of 654 rejected pull requests, from five coding agents and from humans, found that "In our sample, the majority of rejected PRs (67.9%) do not include clear reviewer feedback about why the change was rejected". A study of rejected bug fixes from four agents counts the cost of finding out late: "developers need to review a large amount of changes (a median of 103 lines) to find out that the generated fix is inadequate."

A refusal has three parts: what you found, with the line; why it blocks the merge; and the evidence that would make the change acceptable. The refusal names its evidence, so the next version can meet it, whether the agent writes it or a teammate does. That is also what the week's refusal test asks of you on a seeded generated change: a reason you can point to in the diff, and the evidence that would have made it safe to merge.

*Source: the chapter, section 3.7, with Why Agentic-PRs Get Rejected*

## 2.3 Tests that can fail

*Pull request A adds four tests and they pass. Would any of them fail if the filter were wrong?*

### A test is evidence only if it can fail, and agents edit tests.

**Passes, and is wrong**

- GitHub: "The dangerous one is subtler:"
- "code that compiles, passes every test, and is wrong"
- So ask of each test: could it fail?

**Agents change tests**

- ImpossibleBench edits tests to contradict the task
- GPT-5 cheated 54.0% of the time on those
- Tasks built to be impossible: a rate under pressure

**Two methods it recorded**

- Modify test cases, though told not to
- Special casing: detect the exact test, answer it
- B's edited fixture is the first, in miniature

GitHub's guide names the case that gets through: "The dangerous one is subtler: code that compiles, passes every test, and is wrong."

ImpossibleBench, a benchmark from Carnegie Mellon University and Anthropic, edits tasks' tests so that they contradict the specification, so any pass is a cheat. On its tasks from real repositories with tests made to conflict, "The highest performing model in our tests, GPT-5, cheats 54.0% of the time on Conflicting-SWEbench when facing these clearly impossible tasks." Those are tasks built to be impossible, so the number is a rate of cheating under pressure, not of ordinary work.

Two of the methods it recorded: "Modify Test Cases: The model directly modifies tests despite being explicitly instructed not to." And: "Special Casing: The model special-cases the test cases to pass them. For example, the model may check for the exact test it is facing and return different results based on the test." Pull request B's edited fixture is the first method in miniature.

### Break it on purpose: a break no test catches is a missing test.

<img src="figures/break-the-filter-on-purpose.png" width="864" alt="Left, a frame, Break the filter on purpose, holding three breaks: Return every case; Compare with !=; Match the year as text. An arrow to Run A&#x27;s four tests, after each break, drawn as a stack. From there, two arrows: up to A test fails: it kills the mutant: that test can fail, so it is evidence; and, in orange, down to No test fails: the mutant remains live: a test is missing.">

A study of boundary tests defines the idea: "A test case kills a mutant if it produces a different output on the mutant than on the original program. Otherwise, the mutant remains live."

By hand: make the filter return every case, then make it compare with !=, then make it match the year as text, running the tests after each.

*Source: the chapter, section 3.3, and Adaptive Strategy Generation for Boundary Value Exploration Beyond Numeric Inputs*

### Before you read it, write what must be true, and the evidence.

| What must be true | The evidence |
| --- | --- |
| It does what the prompt asked | A test that fails on main and passes on the branch |
| No test was weakened | Test diff read first: nothing removed, skipped or bent |
| Every name it calls exists | You opened each one |
| No input reaches a query unescaped | A parameterized query; a test with a quotation mark |
| No new dependency is unexplained | The package checked, or the line removed |
| It touches only what the task needed | Every changed file traces to the prompt |
| Someone can explain every line | The author answers the review in their own words |

You can, and doing it first is what keeps the description from setting the standard.

Birgitta Böckeler of Thoughtworks calls those checks computational: "deterministic and fast, run by the CPU. Tests, linters, type checkers, structural analysis. Run in milliseconds to seconds; results are reliable." But no check can supply what nobody stated: "Correctness is outside any sensor's remit if the human didn't clearly specify what they wanted in the first place."

The United States National Institute of Standards and Technology lists eleven minimum verification techniques for developers, among them threat modelling, automated testing, static code scanning, heuristic tools that look for hardcoded secrets, black-box tests, fuzzing and a check of included code such as libraries and packages. NIST presents them as minimum standards, a floor.

For a generated change to the case list, the evidence is this list, and each line is something you can see. In the chapter's full wording: it does what the directing prompt asked, shown by a test that fails on main and passes on the branch, which you ran; no test was weakened, shown by the test diff, read first, with nothing removed, skipped or changed to fit; no input reaches a query or the page unescaped, shown by a parameterized query and a test with a quotation mark in the input.

## 2.4 Ask a generated change for less

*Pull request A works and is tested, and adds things nobody asked for. Is that a reason to say no?*

### Ask for less: keep what traces to the prompt, and nothing else.

<img src="figures/only-what-the-prompt-asked.png" width="864" alt="Pull request A, works, and is tested, above two frames. The scope rule for this course is plain: every file a generated change touches traces to the directing prompt, and anything else goes in its own pull request or goes.

The general filter class is more generic than it needs to be; the configuration option is functionality nobody asked for; the reformatted settings file is a file the task did not need.

*Source: the chapter, section 3.5, with Google's Engineering Practices, What to look for in a code review*

### Speculative code costs even unused: ask for the simpler one.

**Google's reviewer guide**

- Over-engineering: more generic than needed, or a feature not needed now
- "The future problem should be solved once it arrives"
- No other rule applies: stay consistent with the existing code

**What it costs**

- Fowler: the presumptive feature adds complexity
- Complexity makes the software harder to modify and debug
- So every other feature costs more

**For review**

- The Cursor study: AI-generated code needs extra scrutiny
- Ask whether a simpler implementation does the same
- Pull request A: one request for changes

Google's reviewer guide calls it over-engineering, the case "where developers have made the code more generic than it needs to be, or added functionality that isn't presently needed by the system.", and answers it: "The future problem should be solved once it arrives and you can see its actual shape and requirements in the physical universe." On conventions it is as plain: "If no other rule applies, the author should maintain consistency with the existing code."

Martin Fowler counts what speculative code costs even when it is never used: "The code for the presumptive feature adds some complexity to the software, this complexity makes it harder to modify and debug that software, thus increasing the cost of other features."

The study of repositories that adopted Cursor (section 2.5) draws the conclusion for review: "Until these complexities are addressed, software project teams should treat AI-generated code as requiring extra scrutiny during review, with particular attention to whether simpler implementations exist that achieve the same functionality."
