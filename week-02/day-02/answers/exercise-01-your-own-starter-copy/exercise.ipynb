<details><summary>Show the answer: why a template copy, and why Public?</summary>

Use **Use this template**, not **Fork**, and choose **Public**. Why: "You cannot change the visibility of a fork by itself" ([Forks](https://docs.github.com/en/pull-requests/reference/forks)). A fork keeps the starter's whole history. A template copy is a new repository that starts with one commit, and you choose its visibility ([Creating a repository from a template](https://docs.github.com/en/repositories/creating-and-managing-repositories/creating-a-repository-from-a-template)). The check reads your copy without signing in. Git asks for a sign-in only for a private repository.

</details>

## Step 2 · Write your link

This is the solved copy. Run the cell. It builds `practice/` as it is after Step 3, for the made-up account `example-trainee` (simulated). `practice/github.git` stands in for the copy on GitHub, with a new first commit as "Use this template" makes it. The clone's origin is still a github.com link, and Git sends that link to `practice/github.git`, so nothing goes online.

Expect `Wrote practice/starter`, or `Kept` if it is already there.

````python
# Step 2 (solved copy): practice/ as it is after Step 3 (simulated; no network).
# practice/github.git stands in for the trainee's copy on GitHub. The clone keeps a github.com origin;
# a url.<path>.insteadOf setting in the clone sends that address to practice/github.git.
import subprocess
import tempfile
from pathlib import Path

MY_REPO = "https://github.com/example-trainee/ai-full-stack-it-access-requests"
PRACTICE = Path("practice")
STAND_IN = PRACTICE / "github.git"
CLONE = PRACTICE / "starter"
ME = ["-c", "user.name=example-trainee", "-c", "user.email=trainee@example.invalid"]


def git(*args, cwd=None) -> str:
    return subprocess.run(["git", *args], cwd=cwd, check=True, capture_output=True, text=True).stdout


if CLONE.exists():
    print("Kept practice/starter: it is already there.")
else:
    PRACTICE.mkdir(exist_ok=True)
    git("init", "-q", "--bare", "-b", "main", str(STAND_IN))
    # The copy's single new first commit, as Use this template makes it (simulated)
    with tempfile.TemporaryDirectory() as github:
        git("init", "-q", "-b", "main", cwd=github)
        Path(github, "README.md").write_text("# AI Full Stack Challenge — IT Access Requests\n", encoding="utf-8")
        Path(github, "CHALLENGE.md").write_text("# AI Full Stack Challenge — IT Access Requests\n", encoding="utf-8")
        git(*ME, "add", "-A", cwd=github)
        git(*ME, "commit", "-qm", "Initial commit", cwd=github)
        git("push", "-q", str(STAND_IN.resolve()), "main", cwd=github)
    # Step 3: git clone <link> starter, with the link sent to the stand-in
    git("-c", f"url.{STAND_IN.resolve()}.insteadOf={MY_REPO}.git", "clone", "-q", MY_REPO + ".git", str(CLONE))
    git("config", f"url.{STAND_IN.resolve()}.insteadOf", MY_REPO + ".git", cwd=CLONE)
    print("Wrote practice/starter: a clone of the copy, with origin", MY_REPO + ".git", "(simulated)")
````

## Check your work

Run the cell. Expect three `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do. The check reads your copy on GitHub, so it needs the internet.

````python
# Check your work: read practice/starter with Git, then read your copy on GitHub without signing in
import os
import re
import subprocess
from pathlib import Path

CLONE = Path("practice") / "starter"
STARTER_ROOT = "461aac39fc4e6d932fedd1f6e0f76d79eb7e3aad"  # the starter's first commit
NO_SIGN_IN = {key: value for key, value in os.environ.items() if "ASKPASS" not in key}
NO_SIGN_IN["GIT_TERMINAL_PROMPT"] = "0"


def git(*args) -> str:
    """Return git's output in practice/starter, or "" if the command fails. Git never asks to sign in."""
    no_helper = ["-c", "credential.helper=", "-c", "credential.https://github.com.helper="]
    try:
        done = subprocess.run(["git", "-C", str(CLONE), *no_helper, *args], capture_output=True,
                              text=True, timeout=60, env=NO_SIGN_IN)
    except (OSError, subprocess.TimeoutExpired):
        return ""
    return done.stdout.strip() if done.returncode == 0 else ""


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    if not (CLONE / ".git").exists():
        return "practice/starter is missing. Write your link (Step 2), then clone your copy (Step 3)."
    found = re.search(r"github\.com[/:]([\w.-]+)/([\w.-]+)", git("config", "--get", "remote.origin.url"))
    if not found:
        return "practice/starter was not cloned from github.com. Rename it, then clone your copy (Step 3)."
    owner, name = found[1], found[2].removesuffix(".git")
    if owner.lower() == "gitrolltraining":
        return "practice/starter is the course's starter, not your copy. Make your copy (Step 1), then clone it (Step 3)."
    print(f"ok: practice/starter comes from your copy ({owner}/{name})")
    if not git("ls-remote", f"https://github.com/{owner}/{name}.git", "HEAD"):
        return ("Git could not read your copy without signing in. Make it Public on GitHub (Step 1), "
                "or check your internet connection.")
    print("ok: Git reads your copy without signing in, so it is public")
    roots = git("rev-list", "--max-parents=0", "HEAD").split()
    if not roots or STARTER_ROOT in roots:
        return "your copy shares the starter's first commit, so it is a fork. Make a new copy with Use this template (Step 1)."
    print("ok: your copy starts its own history, so it is not a fork")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Do Step 3, then run this cell."
print(f"Not yet: {problem}" if problem else
      "Done: your own public copy of the starter is ready, and its link is the one you hand in.")
````
