## The situation

## Step 1 · Make your copy with "Use this template"

1. Open https://github.com/GitRollTraining/ai-full-stack-it-access-requests.
2. Above the file list, click **Use this template**, then **Create a new repository**.

5. Click **Create repository from template**.

Expect your new repository's page. The address bar shows `https://github.com/<your-account>/<name>`: that is your link.

<details><summary>Show the answer: why a template copy, and why Public?</summary>

Use **Use this template**, not **Fork**, and choose **Public**. Why: "You cannot change the visibility of a fork by itself" ([Forks](https://docs.github.com/en/pull-requests/reference/forks)). A template copy is a new repository that starts with one commit, and you choose its visibility. The course's grading scan has not been able to read private copies. This notebook's check also reads your copy without a sign-in, which works only for public repositories.

</details>

## Step 2 · Save your repository's details

Copy the link from the address bar. Paste it between the quotes in the next cell, then run the cell. It asks GitHub about your repository, with no sign-in, and saves the reply to `practice/repo.json`. Run it again after any change: it writes over the file.

Expect `Saved practice/repo.json` and your account name. If the cell says rate limit, wait and run it again.

```python
# Step 2: paste your repository's link between the quotes, then run this cell
MY_REPO = ""

import json
import re
import urllib.error
import urllib.request
from pathlib import Path

PRACTICE = Path("practice")
PRACTICE.mkdir(exist_ok=True)
found = re.search(r"github\.com/([\w.-]+)/([\w.-]+)", MY_REPO)
if not MY_REPO.strip():
    print("MY_REPO is empty. Paste your link between the quotes, then run this cell again.")
elif not found:
    print(f"{MY_REPO!r} is not a GitHub repository link. It looks like https://github.com/<your-account>/<name>.")
else:
    name = found[2].removesuffix(".git")
    (PRACTICE / "repo.json").unlink(missing_ok=True)  # so the check never reads an older link's reply
    try:
        with urllib.request.urlopen(f"https://api.github.com/repos/{found[1]}/{name}", timeout=20) as reply:
            data = json.load(reply)
    except urllib.error.HTTPError as error:
        if error.code == 404:
            print("GitHub found no public repository at that link. Check the link, and that your copy is Public (Step 1).")
        else:
            print(f"GitHub answered {error.code}: {error.read().decode(errors='replace')[:200]}")
    except OSError as error:
        print(f"Could not reach GitHub ({error}). Check your internet connection, then run this cell again.")
    else:
        (PRACTICE / "repo.json").write_text(json.dumps(data, indent=2), encoding="utf-8")
        print(f"Saved practice/repo.json: {data['full_name']}, owner {data['owner']['login']}")
```

## Check your work

Run the cell. Expect `ok:` lines, then `Done:`. A `Not yet:` line names the first problem and what to do.

```python
# Check your work: read practice/repo.json, which the Step 2 cell saved
import json
from pathlib import Path

PRACTICE = Path("practice")
STARTER = "GitRollTraining/ai-full-stack-it-access-requests"


def first_problem() -> str:
    """Print ok for each check that passes; return the first problem, or ""."""
    path = PRACTICE / "repo.json"
    if not path.is_file():
        return "practice/repo.json is missing. Paste your link in the Step 2 cell and run it."
    try:
        repo = json.loads(path.read_text(encoding="utf-8"))
        owner = repo["owner"]["login"]
    except (ValueError, KeyError, TypeError):
        return "practice/repo.json is not GitHub's reply. Run the Step 2 cell again."
    if owner.lower() == "gitrolltraining":
        return "your link is the course's starter. Make your own copy (Step 1), then save its link (Step 2)."
    print(f"ok: the repository belongs to {owner}")
    if repo.get("private") is not False:
        return "the repository is not public. Make it Public on GitHub (Step 1), then run the Step 2 cell again."
    print("ok: it is public")
    if repo.get("fork") is not False:
        return "the repository is a fork. Make a new copy with Use this template (Step 1), and save its link (Step 2)."
    print("ok: it is not a fork")
    template = repo.get("template_repository")
    if template:
        if template.get("full_name") != STARTER:
            return f"the repository was made from {template.get('full_name')}, not the starter. Do Step 1 again."
        print("ok: it was made from the starter")
    return ""


try:
    problem = first_problem()
except Exception as error:
    problem = f"the check stopped on {type(error).__name__}: {error}. Run the Step 2 cell, then this cell."
print(f"Not yet: {problem}" if problem else
      "Done: your own public copy of the starter is ready, and its link is the one you hand in.")
```
